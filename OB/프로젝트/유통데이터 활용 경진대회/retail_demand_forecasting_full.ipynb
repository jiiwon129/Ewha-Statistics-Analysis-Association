### Purchase (매입 데이터) 불러오기 및 파생 변수 생성

```python
# 라이브러리 설치
import pandas as pd
import holidays

# 데이터 불러오기
df = pd.read_csv('A_purchase_1.csv')

# 시간 피처 생성
# day 컬럼 datetime 변환
df['date'] = pd.to_datetime(df['date'], errors='coerce')

# 한국 공휴일 세팅 (데이터 연도 자동 포함)
kr_holidays = holidays.KR(years=df['date'].dt.year.unique())

# holiday: 주말 또는 공휴일이면 1, 아니면 0
df['holiday'] = df['date'].apply(
    lambda x: 1 if (x.weekday() >= 5 or x in kr_holidays) else 0
)

# quarter: 월 기준 1~4분기
df['quarter'] = df['date'].dt.month.map({
    1:1, 2:1, 3:1,
    4:2, 5:2, 6:2,
    7:3, 8:3, 9:3,
    10:4, 11:4, 12:4
})

# 드롭할 컬럼 리스트
drop_cols = [
    "client_id", "client_zipcode", "supplier_code", "supplier_zipcode", "stock_in_type", "barcode",
    "spec", "option_code", "option_name", "units_per_pack",
    "each_count", "vat", "main_category", "sub_category"
]

# 컬럼 삭제
df = df.drop(columns=drop_cols, errors="ignore")
```

```python
# 과거 데이터 기반 피처 생성
import pandas as pd
import numpy as np

# 설정
PURCHASE_DATE_COL = "date"
PRODUCT_KEY_COLS  = ["product_code"]
QTY_COL           = "quantity"
PRICE_COL         = "sales_price"     # 매입 금액(총액)

def _ensure_datetime(df, date_col):
    df = df.copy()
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
    return df

def _make_product_key(df, key_cols):
    if len(key_cols) == 1:
        return df[key_cols[0]].astype(str)
    return df[key_cols].astype(str).agg("§".join, axis=1)

# 기본 전처리
df = _ensure_datetime(df, PURCHASE_DATE_COL).copy()
df["_prod_key"] = _make_product_key(df, PRODUCT_KEY_COLS)
df["_ym"] = df[PURCHASE_DATE_COL].dt.to_period("M").astype(str)

# '입고' 월별 수량 집계 (work_type에 '입고' 포함 가정)
if "work_type" in df.columns:
    inbound_monthly = (
        df.loc[df["work_type"].astype(str).str.contains("입고", na=False)]
          .groupby(["_prod_key", "_ym"], as_index=False)[QTY_COL].sum()
          .rename(columns={QTY_COL: "inbound_qty"})
    )
else:
    inbound_monthly = (
        df.groupby(["_prod_key", "_ym"], as_index=False)[QTY_COL].sum()
          .rename(columns={QTY_COL: "inbound_qty"})
    )

# 시계열 정렬
inbound_monthly["_ym_period"] = pd.to_datetime(inbound_monthly["_ym"] + "-01")
inbound_monthly = inbound_monthly.sort_values(["_prod_key", "_ym_period"])

# 지연/롤링(3개월, 현재월 포함)
def _inbound_lag_roll(g, window=3):
    g = g.copy()
    g["inbound_qty"] = g["inbound_qty"].fillna(0)
    g["inbound_qty_lag_1m"] = g["inbound_qty"].shift(1)
    g["inbound_qty_lag_1y"] = g["inbound_qty"].shift(12)

    g["inbound_rolling_avg_3m"] = g["inbound_qty"].rolling(window, min_periods=1).mean()
    g["inbound_rolling_max_3m"] = g["inbound_qty"].rolling(window, min_periods=1).max()
    g["inbound_rolling_min_3m"] = g["inbound_qty"].rolling(window, min_periods=1).min()
    return g

inbound_monthly = inbound_monthly.groupby("_prod_key", group_keys=False).apply(_inbound_lag_roll, window=3)

# 가격 피처 (unit_price, price_volatility 3개월 표준편차)
df["_safe_qty"] = df[QTY_COL].replace(0, np.nan)
df["unit_price_row"] = df[PRICE_COL] / df["_safe_qty"]
price_monthly = (
    df.groupby(["_prod_key", "_ym"], as_index=False)["unit_price_row"].mean()
      .rename(columns={"unit_price_row": "unit_price"})
)
price_monthly["_ym_period"] = pd.to_datetime(price_monthly["_ym"] + "-01")
price_monthly = price_monthly.sort_values(["_prod_key", "_ym_period"])

def _price_vol(g, window=3):
    g = g.copy()
    g["price_volatility"] = g["unit_price"].rolling(window=window, min_periods=2).std()
    return g

price_monthly = price_monthly.groupby("_prod_key", group_keys=False).apply(_price_vol, window=3)

# 원본 df에 병합 (제품, 월 기준)
df = df.merge(
    inbound_monthly[["_prod_key", "_ym",
                     "inbound_qty_lag_1m", "inbound_qty_lag_1y",
                     "inbound_rolling_avg_3m", "inbound_rolling_max_3m", "inbound_rolling_min_3m"]],
    on=["_prod_key", "_ym"], how="left"
)
df = df.merge(
    price_monthly[["_prod_key", "_ym", "unit_price", "price_volatility"]],
    on=["_prod_key", "_ym"], how="left"
)

# 보조 컬럼 제거
df = df.drop(columns=["_prod_key", "_ym", "_ym_period", "_safe_qty", "unit_price_row"], errors="ignore")
```

```python
# 날씨 데이터, 소비자 물가지수, 소비자 심리지수

# 외부 데이터 불러오기
weather = pd.read_csv("weather_data.csv")
cpi  = pd.read_csv("소비자물가지수.csv")
ccsi = pd.read_csv("소비자심리지수.csv", encoding='cp949')

# CPI
cpi = pd.melt(cpi, id_vars=['시도별'], var_name='date', value_name='cpi')
cpi['date'] = pd.to_datetime(cpi['date'], format='%Y.%m')
cpi['year'] = cpi['date'].dt.year
cpi['month'] = cpi['date'].dt.month
cpi = cpi.drop(columns=["date"])

# CCSI
ccsi['시점'] = pd.to_datetime(ccsi['시점'], format='%Y.%m 월')
ccsi = ccsi[ccsi['CSI코드별'] == '소비자심리지수'].iloc[:,[2, 3]]
ccsi.columns = ['date', 'ccsi']
ccsi['year'] = ccsi['date'].dt.year
ccsi['month'] = ccsi['date'].dt.month
ccsi = ccsi.drop(columns=["date"])

# df (매입 데이터) 처리
# 날짜 -> datetime 변환
df["date"] = pd.to_datetime(df["date"], errors="coerce")

# year, month 추출
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month

# 조인
df = df.merge(weather, on=["year", "month"], how="left")
df = df.merge(cpi,     on=["year", "month"], how="left")
df = df.merge(ccsi,    on=["year", "month"], how="left")

# zero sugar 변수 추가
# "제로"라는 문자열이 product_name에 있으면 1, 없으면 0
df["zero_sugar"] = df["product_name"].str.contains("제로", na=False).astype(int)

# 지연변수 생성(2025.01-2025.03)
def make_q4_ref_and_future(
    data: pd.DataFrame,
    key_col=["product_name"],
    year_col="year",
    month_col="month",
    target_cols=("quantity", "sales_price")
):
    df = data.copy()

    # 1) Q4 기준 레퍼런스 만들기 ----
    df["yyyymm"] = df[year_col].astype(int) * 100 + df[month_col].astype(int)
    q4 = df[(df[year_col] == 2024) & (df[month_col].isin([10, 11, 12]))].copy()
    q4 = q4.sort_values([key_col, "yyyymm"])

    time_cols = [c for c in [year_col, month_col, "day", "yyyymm"] if c in df.columns]
    exclude_cols = set([key_col, *time_cols, *target_cols])
    cols_to_uniform = [c for c in df.columns if c not in exclude_cols]

    # 품목별 Q4 최신값(결측은 ffill 후 마지막 행) 추출
    # Q4에서 각 품목별 가장 최신 row
    q4_latest = (
        q4.sort_values(["product_name", "yyyymm"])
          .groupby("product_name")
          .last()
          .reset_index()
    )

    # 통일할 컬럼만 선택
    ref = q4_latest[[key_col] + cols_to_uniform]


    # 2) 전체 데이터에 레퍼런스 머지(속성 통일) ----
    merged = df.merge(ref, on=key_col, how="left", suffixes=("", "_q4ref"))
    for c in cols_to_uniform:
        merged[c] = merged[f"{c}_q4ref"].combine_first(merged[c])
        merged.drop(columns=[f"{c}_q4ref"], inplace=True)
    merged.drop(columns=["yyyymm"], errors="ignore", inplace=True)
    df_uniform = merged

    # 3) 2025.01/02/03 스켈레톤 생성 + Q4 속성 부여 ----
    items_2025 = ref[key_col].drop_duplicates()
    future_idx = pd.MultiIndex.from_product(
        [[2025], [1, 2, 3], items_2025],
        names=[year_col, month_col, key_col]
    )
    future = future_idx.to_frame(index=False)

    # Q4 속성 붙이기
    future = future.merge(ref, on=key_col, how="left")

    # 예측 대상(y) 초기화
    for c in target_cols:
        if c in df.columns:
            future[c] = np.nan

    # 있으면 day 기본값 1로 세팅
    if "day" in df.columns:
        future["day"] = 1

    future_2025 = future

    return df_uniform, future_2025

df_uniform, future_2025 = make_q4_ref_and_future(df, key_col="product_name")

# holiday_count 파생 변수 생성
def add_holiday_count(df, year_col="year", month_col="month", holiday_col="holiday"):
    """
    같은 (year, month) 기준으로 holiday 합계를 구해서 holiday_count 파생변수 생성
    """
    # (1) 연월별 holiday 합계 계산
    holiday_sum = (
        df.groupby([year_col, month_col])[holiday_col]
          .sum()
          .reset_index()
          .rename(columns={holiday_col: "holiday_count"})
    )

    # (2) 원본 df에 merge
    df = df.merge(holiday_sum, on=[year_col, month_col], how="left")

    return df

# df에 적용
df = add_holiday_count(df)
```

```python
import pandas as pd
import numpy as np

# 설정
SALES_DATE_COL   = "sales_date"
PRODUCT_KEY_COLS = ["barcode"]
QTY_COL          = "quantity"
SUPPLY_COL       = "supply_price"    # 매출 공급가(총액)

# 기본 전처리
df2 = _ensure_datetime(df2, SALES_DATE_COL).copy()
df2["_prod_key"] = _make_product_key(df2, PRODUCT_KEY_COLS)
df2["_ym"] = df2[SALES_DATE_COL].dt.to_period("M").astype(str)

# 판매 월별 수량 집계
sales_monthly = (
    df2.groupby(["_prod_key", "_ym"], as_index=False)[QTY_COL].sum()
       .rename(columns={QTY_COL: "sales_qty"})
)
sales_monthly["_ym_period"] = pd.to_datetime(sales_monthly["_ym"] + "-01")
sales_monthly = sales_monthly.sort_values(["_prod_key", "_ym_period"])

# 지연/롤링(3개월, 현재월 포함)
def _sales_lag_roll(g, window=3):
    g = g.copy()
    g["sales_qty"] = g["sales_qty"].fillna(0)
    g["sales_qty_lag_1m"] = g["sales_qty"].shift(1)
    g["sales_qty_lag_1y"] = g["sales_qty"].shift(12)

    g["sales_rolling_avg_3m"] = g["sales_qty"].rolling(window, min_periods=1).mean()
    g["sales_rolling_max_3m"] = g["sales_qty"].rolling(window, min_periods=1).max()
    g["sales_rolling_min_3m"] = g["sales_qty"].rolling(window, min_periods=1).min()
    return g

sales_monthly = sales_monthly.groupby("_prod_key", group_keys=False).apply(_sales_lag_roll, window=3)

# 가격 피처 (unit_supply_price, supply_price_volatility 3개월 표준편차)
df2["_safe_qty"] = df2[QTY_COL].replace(0, np.nan)
df2["unit_supply_price_row"] = df2[SUPPLY_COL] / df2["_safe_qty"]
supply_price_monthly = (
    df2.groupby(["_prod_key", "_ym"], as_index=False)["unit_supply_price_row"].mean()
       .rename(columns={"unit_supply_price_row": "unit_supply_price"})
)
supply_price_monthly["_ym_period"] = pd.to_datetime(supply_price_monthly["_ym"] + "-01")
supply_price_monthly = supply_price_monthly.sort_values(["_prod_key", "_ym_period"])

def _supply_vol(g, window=3):
    g = g.copy()
    g["supply_price_volatility"] = g["unit_supply_price"].rolling(window=window, min_periods=2).std()
    return g

supply_price_monthly = supply_price_monthly.groupby("_prod_key", group_keys=False).apply(_supply_vol, window=3)

# 원본 df2에 병합 (제품, 월 기준)
df2 = df2.merge(
    sales_monthly[["_prod_key", "_ym",
                   "sales_qty_lag_1m", "sales_qty_lag_1y",
                   "sales_rolling_avg_3m", "sales_rolling_max_3m", "sales_rolling_min_3m"]],
    on=["_prod_key", "_ym"], how="left"
)
df2 = df2.merge(
    supply_price_monthly[["_prod_key", "_ym", "unit_supply_price", "supply_price_volatility"]],
    on=["_prod_key", "_ym"], how="left"
)

# 보조 컬럼 제거
df2 = df2.drop(columns=["_prod_key", "_ym", "_ym_period", "_safe_qty", "unit_supply_price_row"], errors="ignore")

# df2 (매출 데이터) 처리
# 날짜 -> datetime 변환
df2["sales_date"] = pd.to_datetime(df2["sales_date"], errors="coerce")

# year, month 추출
df2["year"] = df2["sales_date"].dt.year
df2["month"] = df2["sales_date"].dt.month

# 조인
df2 = df2.merge(weather, on=["year", "month"], how="left")
df2 = df2.merge(cpi,     on=["year", "month"], how="left")
df2 = df2.merge(ccsi,    on=["year", "month"], how="left")

# zero sugar 변수 추가
df2["zero_sugar"] = df2["product_name"].str.contains("제로", na=False).astype(int)

# 지연변수 생성
df2_uniform, future2_2025 = make_q4_ref_and_future(df2, key_col="product_name")

# holiday_count 파생 변수 적용
df2 = add_holiday_count(df2)
```

### 모델링

```python
import pandas as pd
from pandas.api.types import (
    is_datetime64_any_dtype,
    is_datetime64tz_dtype,
    is_timedelta64_dtype,
    is_period_dtype,
    is_object_dtype,
)

def find_datetime_like_cols(df, name="df", detect_object_like=True, object_like_threshold=0.9):
    """
    - native datetime/tz/timedelta/period dtype을 탐지
    - (옵션) object 컬럼 중 '대부분'이 날짜로 해석되는 열도 탐지
    """
    native_dt = []
    object_dt_like = []

    for c in df.columns:
        s = df[c]
        if (is_datetime64_any_dtype(s) or
            is_datetime64tz_dtype(s) or
            is_timedelta64_dtype(s) or
            is_period_dtype(s)):
            native_dt.append(c)
        elif detect_object_like and is_object_dtype(s):
            # object 열이지만 날짜처럼 보이는지 휴리스틱으로 확인
            conv = pd.to_datetime(s, errors="coerce", infer_datetime_format=True)
            if conv.notna().mean() >= object_like_threshold:
                object_dt_like.append(c)

    print(f"[{name}] native datetime-like:", native_dt if native_dt else "없음")
    if detect_object_like:
        print(f"[{name}] object→datetime 후보:", object_dt_like if object_dt_like else "없음")
    return native_dt, object_dt_like

# 실행
_ = find_datetime_like_cols(df, "df")
_ = find_datetime_like_cols(df2, "df2")

# datetime 열 제거
df = df.drop(columns=["date"], errors="ignore")
df2 = df2.drop(columns=["sales_date"], errors="ignore")

# 인코딩
from sklearn.preprocessing import LabelEncoder

def encode_categoricals(df, cat_cols):
    """
    주어진 df의 cat_cols 열들을 Label Encoding 해서 새로운 df 반환
    """
    df = df.copy()
    for col in cat_cols:
        le = LabelEncoder()
        df[col] = df[col].astype(str).fillna("NA")   # NaN -> "NA"
        df[col] = le.fit_transform(df[col])
    return df

# df 인코딩
df = encode_categoricals(df, ["work_type", "product_name", "mid_category"])

# df2 인코딩
df2 = encode_categoricals(df2, ["type", "product_name", "mid_category"])

# 모델 및 성능 지표 선언
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.ensemble import RandomForestRegressor
try:
    from lightgbm import LGBMRegressor
except ImportError:
    LGBMRegressor = None
try:
    from xgboost import XGBRegressor
except ImportError:
    XGBRegressor = None

# CONFIG 설정
RANDOM_STATE = 42
TARGET_COL = "quantity"
YEAR_COL   = "year"
MONTH_COL  = "month"

# Utils
def ensure_year_month(df, year_col="year", month_col="month"):
    """year/month 컬럼이 사라졌을 경우 다양한 소스에서 복구"""
    df = df.copy()

    # 인덱스 레벨에 있으면 컬럼으로 빼오기
    if isinstance(df.index, pd.MultiIndex):
        idx_names = list(df.index.names)
        if year_col in idx_names or month_col in idx_names:
            df = df.reset_index()
    elif df.index.name in [year_col, month_col]:
        df = df.reset_index()

    # 병합 후 year_x/year_y -> year 로 표준화
    cand_ys = [c for c in df.columns if c.lower().startswith("year")]
    cand_ms = [c for c in df.columns if c.lower().startswith("month")]
    if year_col not in df.columns and len(cand_ys) > 0:
        df.rename(columns={cand_ys[0]: year_col}, inplace=True)
    if month_col not in df.columns and len(cand_ms) > 0:
        df.rename(columns={cand_ms[0]: month_col}, inplace=True)

    # 이미 있으면 통과
    if year_col in df.columns and month_col in df.columns:
        pass
    elif "yyyymm" in df.columns:
        y = pd.to_numeric(df["yyyymm"], errors="coerce")
        df[year_col]  = (y // 100).astype("Int64")
        df[month_col] = (y % 100).astype("Int64")
    elif "date" in df.columns:
        d = pd.to_datetime(df["date"], errors="coerce")
        df[year_col], df[month_col] = d.dt.year, d.dt.month
    elif "sales_date" in df.columns:
        d = pd.to_datetime(df["sales_date"], errors="coerce")
        df[year_col], df[month_col] = d.dt.year, d.dt.month
    else:
        raise KeyError(f"{year_col}, {month_col} not found in DataFrame")

    # 숫자/범위/결측 보정
    df[year_col]  = pd.to_numeric(df[year_col], errors="coerce").astype("Int64")
    df[month_col] = pd.to_numeric(df[month_col], errors="coerce").astype("Int64")
    df[month_col] = df[month_col].clip(1, 12)

    miss = df[year_col].isna() | df[month_col].isna()
    if miss.any():
        print(f"[warn] drop rows missing {year_col}/{month_col} after ensure_year_month: {int(miss.sum())}")
        df = df.loc[~miss].copy()

    df[year_col]  = df[year_col].astype(int)
    df[month_col] = df[month_col].astype(int)
    return df


def assert_has_columns(df, cols, where=""):
    """필수 컬럼이 없으면 KeyError 발생"""
    missing = [c for c in cols if c not in df.columns]
    if missing:
        raise KeyError(f"Missing columns {missing} in {where}.\nCurrent columns: {list(df.columns)[:30]} ...")

# Feature Engineering
def add_time_key(df, id_col):
    df = ensure_year_month(df, YEAR_COL, MONTH_COL)
    assert_has_columns(df, [id_col, YEAR_COL, MONTH_COL], "add_time_key(input)")

    df = df.copy()
    df[[YEAR_COL, MONTH_COL]] = df[[YEAR_COL, MONTH_COL]].replace([np.inf, -np.inf], np.nan)
    df[YEAR_COL]  = pd.to_numeric(df[YEAR_COL],  errors="coerce")
    df[MONTH_COL] = pd.to_numeric(df[MONTH_COL], errors="coerce")

    if id_col in df.columns:
        df[YEAR_COL]  = df.groupby(id_col)[YEAR_COL].transform(lambda s: s.ffill().bfill())
        df[MONTH_COL] = df.groupby(id_col)[MONTH_COL].transform(lambda s: s.ffill().bfill())

    df[MONTH_COL] = df[MONTH_COL].round().clip(1, 12)

    miss = df[YEAR_COL].isna() | df[MONTH_COL].isna()
    if miss.any():
        print(f"[warn] drop rows with missing year/month after cleaning: {int(miss.sum())}")
        df = df.loc[~miss].copy()

    df[YEAR_COL]  = df[YEAR_COL].astype(int)
    df[MONTH_COL] = df[MONTH_COL].astype(int)
    df["yyyymm"] = df[YEAR_COL] * 100 + df[MONTH_COL]

    df.sort_values([id_col, "yyyymm"], inplace=True)
    return df


def build_features(df, id_col, lags=(1,3,6,12), roll_windows=(3,6)):
    df = add_time_key(df, id_col)
    g = df.groupby(id_col, group_keys=False)

    for L in lags:
        df[f"lag_{L}"] = g[TARGET_COL].shift(L)

    for W in roll_windows:
        df[f"roll_mean_{W}"] = g[TARGET_COL].shift(1).rolling(W, min_periods=1).mean()
        df[f"roll_std_{W}"]  = g[TARGET_COL].shift(1).rolling(W, min_periods=1).std()

    df["month_sin"] = np.sin(2*np.pi*(df[MONTH_COL]/12))
    df["month_cos"] = np.cos(2*np.pi*(df[MONTH_COL]/12))
    df["quarter"]   = ((df[MONTH_COL]-1)//3 + 1)
    return df


def get_feature_cols(df, id_col):
    drop = {TARGET_COL, YEAR_COL, MONTH_COL, "day", "yyyymm", id_col}
    return [c for c in df.columns if c not in drop]


def to_numeric_matrix(df, feature_cols):
    X = df[feature_cols].copy()
    X = X.select_dtypes(include=[np.number]).replace([np.inf, -np.inf], np.nan).fillna(0)
    return X

# Split
def split_train_val(df, id_col):
    df = build_features(df, id_col)
    assert_has_columns(df, [id_col, YEAR_COL, MONTH_COL, "yyyymm"], "split_train_val(after build_features)")

    need_lags = [c for c in df.columns if c.startswith("lag_")]
    train = df[df[YEAR_COL] <= 2023].dropna(subset=[TARGET_COL] + need_lags)
    val   = df[df[YEAR_COL] == 2024].dropna(subset=[TARGET_COL] + need_lags)

    feat_cols = sorted(list(set(get_feature_cols(train, id_col)).intersection(get_feature_cols(val, id_col))))
    X_tr = to_numeric_matrix(train, feat_cols)
    X_vl = to_numeric_matrix(val,   feat_cols)
    y_tr, y_vl = train[TARGET_COL], val[TARGET_COL]
    return df, X_tr, y_tr, X_vl, y_vl

# Models
def get_models(random_state: int = RANDOM_STATE):
    models = {
        "rf": RandomForestRegressor(
            n_estimators=400, max_depth=12, random_state=random_state, n_jobs=-1
        )
    }
    if LGBMRegressor is not None:
        models["lgbm"] = LGBMRegressor(
            n_estimators=600, learning_rate=0.05,
            subsample=0.9, colsample_bytree=0.9,
            random_state=random_state,
        )
    if XGBRegressor is not None:
        models["xgb"] = XGBRegressor(
            n_estimators=700, learning_rate=0.05,
            max_depth=7, subsample=0.9, colsample_bytree=0.9,
            random_state=random_state, tree_method="hist"
        )
    return models

# Evaluate
def evaluate_models(df, id_col):
    df, X_tr, y_tr, X_vl, y_vl = split_train_val(df, id_col)
    models = get_models()
    results = {}
    for name, mdl in models.items():
        mdl.fit(X_tr, y_tr)
        pred = mdl.predict(X_vl)
        rmse = np.sqrt(mean_squared_error(y_vl, pred))
        mae  = mean_absolute_error(y_vl, pred)
        results[name] = {"RMSE": rmse, "MAE": mae, "model": mdl}
        print(f"[{name}] RMSE={rmse:.3f} | MAE={mae:.3f}")
    return df, results


def dedupe_cols(df: pd.DataFrame) -> pd.DataFrame:
    return df.loc[:, ~df.columns.duplicated()].copy()

# Future skeleton
def make_future_frame(df_base, id_col):
    base = add_time_key(df_base.copy(), id_col)
    assert_has_columns(base, [id_col, YEAR_COL, MONTH_COL, "yyyymm"], "make_future_frame(base)")

    last_meta = (base.sort_values([id_col, "yyyymm"])
                      .groupby(id_col, as_index=False).last())

    n = len(last_meta)
    future = pd.DataFrame({
        YEAR_COL:  np.repeat(2025, n*3).astype(int),
        MONTH_COL: np.tile([1,2,3], n).astype(int),
        id_col:    np.repeat(last_meta[id_col].values, 3)
    })

    # 여기서 year/month도 제외해야 중복이 안 생깁니다
    drop_set = {id_col, TARGET_COL, "yyyymm", "day", YEAR_COL, MONTH_COL}
    keep = [c for c in last_meta.columns
            if c not in drop_set
            and not c.startswith("lag_")
            and not c.startswith("roll_")]

    future = future.merge(
        last_meta[[id_col] + keep],
        on=id_col, how="left", validate="many_to_one"
    )
    future[TARGET_COL] = np.nan

    # 중복 컬럼 제거 가드
    future = dedupe_cols(future)
    future = ensure_year_month(future, YEAR_COL, MONTH_COL)
    assert_has_columns(future, [id_col, YEAR_COL, MONTH_COL], "make_future_frame(future)")

    return future[[id_col, YEAR_COL, MONTH_COL] + keep + [TARGET_COL]]

# Retrain & Predict
def retrain_and_predict_2025(df, id_col, best_model_name="lgbm", strategy="direct", verbose=True):
    df_all = build_features(df.copy(), id_col)
    assert_has_columns(df_all, [id_col, YEAR_COL, MONTH_COL, "yyyymm"], "retrain(df_all)")

    need_lags = [c for c in df_all.columns if c.startswith("lag_")]
    train_all = df_all[df_all[YEAR_COL] <= 2024].dropna(subset=[TARGET_COL] + need_lags)

    feat_cols = get_feature_cols(train_all, id_col)
    X_all = to_numeric_matrix(train_all, feat_cols)
    y_all = train_all[TARGET_COL]

    models = get_models()
    if best_model_name not in models:
        best_model_name = list(models.keys())[0]
    mdl = models[best_model_name]
    mdl.fit(X_all, y_all)

    future = make_future_frame(df_all, id_col)
    if verbose:
        print(f"[debug] future rows: {len(future)} | unique items: {future[id_col].nunique()}")

    merged = pd.concat([df_all, future], ignore_index=True)
    merged = build_features(merged, id_col)
    merged = dedupe_cols(merged)
    merged = ensure_year_month(merged, YEAR_COL, MONTH_COL)
    assert_has_columns(merged, [id_col, YEAR_COL, MONTH_COL, "yyyymm"], "retrain(merged)")

    merged["yyyymm"] = pd.to_numeric(merged["yyyymm"], errors="coerce").astype("Int64")
    mask_2025 = merged["yyyymm"].between(202501, 202503)
    feat_all = get_feature_cols(merged, id_col)

    if strategy == "direct":
        if verbose:
            print(f"[debug] direct rows:", int(mask_2025.sum()), "| feat_all:", len(feat_all))
        X_25 = to_numeric_matrix(merged.loc[mask_2025], feat_all)
        out = merged.loc[mask_2025, [id_col, YEAR_COL, MONTH_COL]].copy()
        out["pred_quantity"] = mdl.predict(X_25) if X_25.shape[0] else np.nan
        return out.sort_values([id_col, YEAR_COL, MONTH_COL])

    elif strategy == "recursive":
        out_list = []
        for ym in [202501, 202502, 202503]:
            m = (merged["yyyymm"] == ym)
            if verbose:
                print(f"[debug] recursive ym={ym} rows:", int(m.sum()))
            X_m = to_numeric_matrix(merged.loc[m], feat_all)
            if X_m.shape[0] == 0:
                tmp = merged.loc[m, [id_col, YEAR_COL, MONTH_COL]].copy()
                tmp["pred_quantity"] = np.nan
                out_list.append(tmp); continue
            y_m = mdl.predict(X_m)
            tmp = merged.loc[m, [id_col, YEAR_COL, MONTH_COL]].copy()
            tmp["pred_quantity"] = y_m
            out_list.append(tmp)
            merged.loc[m, TARGET_COL] = y_m
            merged = build_features(merged, id_col)
        return (pd.concat(out_list, ignore_index=True)
                if out_list else merged.loc[mask_2025, [id_col, YEAR_COL, MONTH_COL]].assign(pred_quantity=np.nan)
               ).sort_values([id_col, YEAR_COL, MONTH_COL])
```

```python
# 1) df (id_col = product_code)
df_built, results_df = evaluate_models(df, id_col="product_code")
best_name_df = min(results_df, key=lambda k: results_df[k]["RMSE"])
print("[df] Best model:", best_name_df, results_df[best_name_df])

pred_2025_recursive_df = retrain_and_predict_2025(
    df_built, id_col="product_code",
    best_model_name=best_name_df, strategy="recursive", verbose=True
)

# 2) df2 (id_col = barcode)
df2_built, results_df2 = evaluate_models(df2, id_col="barcode")
best_name_df2 = min(results_df2, key=lambda k: results_df2[k]["RMSE"])
print("[df2] Best model:", best_name_df2, results_df2[best_name_df2])

pred2_2025_recursive = retrain_and_predict_2025(
    df2_built, id_col="barcode",
    best_model_name=best_name_df2, strategy="recursive", verbose=True
)

# 최종 target 확인
def aggregate_pred_df(pred_df, label="매입"):
    """
    pred_df: (id_col, year, month, pred_quantity) 컬럼을 가진 예측 결과 DF
    label  : '매입' / '매출' 등 표시용 텍스트
    반환   : 1행짜리 wide DF (예: 25년 1월 매입, 25년 2월 매입, 25년 3월 매입)
    """
    if pred_df is None or len(pred_df) == 0:
        # 빈 DF 방어
        cols = [f"25년 {m}월 {label}" for m in [1, 2, 3]]
        return pd.DataFrame([[np.nan, np.nan, np.nan]], columns=cols)

    # 1) 2025-01~03만 집계
    tmp = (pred_df
           .loc[(pred_df["year"] == 2025) & (pred_df["month"].isin([1, 2, 3]))]
           .groupby(["year", "month"], as_index=False)["pred_quantity"].sum())

    # 2) 컬럼명 생성 (25년 1월 매입 등)
    tmp["colname"] = tmp["year"].mod(100).astype(int).astype(str) + "년 " \
                     + tmp["month"].astype(int).astype(str) + "월 " + label

    # 3) 1행 딕셔너리로 변환 후, 월 순서대로 정렬(reindex)
    desired_cols = [f"25년 {m}월 {label}" for m in [1, 2, 3]]
    row = tmp.set_index("colname")["pred_quantity"].reindex(desired_cols, fill_value=0)

    return pd.DataFrame([row.values], columns=desired_cols)

agg_df_recursive = aggregate_pred_df(pred_2025_recursive_df, label="매입")
display(agg_df_recursive)

agg_df2_recursive = aggregate_pred_df(pred2_2025_recursive, label="매출")
display(agg_df2_recursive)
```

### 2. 과거 데이터 기반 피처
#### 지연값 데이터
- sales_qty_lag_1m, inbound_qty_lag_1m: 1개월 전 동일 상품의 판매수량 및 입고수량
- sales_qty_lag_1y, inbound_qty_lag_1y: 1년 전 동일 월의 판매수량 및 입고수량

```python
import pandas as pd

# ── 1) sales: 월 집계 만들고 lag 생성
sales = sales.copy()

# sales_date가 일단위면 월로 내림
sales['year'] = pd.to_datetime(sales['sales_date']).dt.year
sales['month'] = pd.to_datetime(sales['sales_date']).dt.month

sales_monthly = (
    sales.groupby(['product_name','year','month'], as_index=False)['quantity']
         .sum()
         .rename(columns={'quantity':'sales_qty'})
)

# shift 전에 정렬이 필수!
sales_monthly = sales_monthly.sort_values(['product_name','year','month'])

# lag 1개월 / 12개월
sales_monthly['sales_qty_lag_1m'] = sales_monthly.groupby('product_name')['sales_qty'].shift(1)
sales_monthly['sales_qty_lag_1y'] = sales_monthly.groupby('product_name')['sales_qty'].shift(12)

# ── 2) purchase: 월 집계 만들고 lag 생성
purchase = purchase.copy()

# purchase의 날짜 컬럼명이 'date' 라면 아래처럼 월 추출
purchase['year'] = pd.to_datetime(purchase['date']).dt.year
purchase['month'] = pd.to_datetime(purchase['date']).dt.month

purchase_monthly = (
    purchase.groupby(['product_name','year','month'], as_index=False)['quantity']
            .sum()
            .rename(columns={'quantity':'inbound_qty'})
)

purchase_monthly = purchase_monthly.sort_values(['product_name','year','month'])
purchase_monthly['inbound_qty_lag_1m'] = purchase_monthly.groupby('product_name')['inbound_qty'].shift(1)
purchase_monthly['inbound_qty_lag_1y'] = purchase_monthly.groupby('product_name')['inbound_qty'].shift(12)

# ── 3) 원본 DF에 붙이기 (행 단위 데이터에 월 기준으로 매핑)
# 원본이 일단위라면 year, month 키로 merge해서 같은 월의 lag 값을 모두 같은 월 행에 복제해 붙입니다.
sales = sales.merge(
    sales_monthly[['product_name','year','month','sales_qty_lag_1m','sales_qty_lag_1y']],
    on=['product_name','year','month'], how='left'
)

purchase = purchase.merge(
    purchase_monthly[['product_name','year','month','inbound_qty_lag_1m','inbound_qty_lag_1y']],
    on=['product_name','year','month'], how='left'
)

# ── 4) 확인
print("sales cols:", sales.columns.tolist())
print("purchase cols:", purchase.columns.tolist())
print(sales[['product_name','year','month','sales_qty_lag_1m','sales_qty_lag_1y']].head())
print(purchase[['product_name','year','month','inbound_qty_lag_1m','inbound_qty_lag_1y']].head())
```

### 1. sales 데이터모델링




```python
# === Single-cell pipeline: 2021~2023 Train / 2024 Val / 2025Q1 Test ===
import pandas as pd, numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.ensemble import RandomForestRegressor
import lightgbm as lgb

# (옵션) XGBoost가 없다면 자동 스킵
try:
    from xgboost import XGBRegressor
    HAS_XGB = True
except Exception:
    HAS_XGB = False

RANDOM_STATE = 42

# -----------------------------
# 0) 기본 정리
# -----------------------------
assert {"sales_date","product_name","quantity"}.issubset(sales.columns), "필수 컬럼 누락"
raw = sales.copy()
raw["sales_date"] = pd.to_datetime(raw["sales_date"], errors="coerce")
raw = raw.dropna(subset=["sales_date"])

# -----------------------------
# 1) 월 집계 (제품×월)
# -----------------------------
raw["ym"] = raw["sales_date"].dt.to_period("M").dt.to_timestamp()
agg = {"quantity":"sum"}
if "unit_supply_price" in raw.columns: agg["unit_supply_price"] = "mean"

monthly_all = (raw.groupby(["product_name","ym"])
                 .agg(**{k:(k,v) for k,v in agg.items()})
                 .reset_index())
monthly_all.columns = ["product_name","ym"] + list(agg.keys())
monthly_all = monthly_all.sort_values(["product_name","ym"]).reset_index(drop=True)

# product_name을 LabelEncode → 모든 모델에 공통 사용 가능
le = LabelEncoder()
monthly_all["product_id"] = le.fit_transform(monthly_all["product_name"])

# -----------------------------
# 2) 피처 생성 함수 (미래 누수 방지: 모두 shift(1))
# -----------------------------
def build_features(df):
    df = df.sort_values(["product_name","ym"]).copy()

    gq = df.groupby("product_name")["quantity"]
    df["qty_lag_1m"]      = gq.shift(1)
    df["qty_lag_3m_mean"] = gq.shift(1).rolling(3).mean().reset_index(level=0, drop=True)
    df["qty_lag_3m_max"]  = gq.shift(1).rolling(3).max().reset_index(level=0, drop=True)
    df["qty_lag_3m_min"]  = gq.shift(1).rolling(3).min().reset_index(level=0, drop=True)
    df["qty_lag_12m"]     = gq.shift(12)

    if "unit_supply_price" in df.columns:
        df["price_volatility"] = (df.groupby("product_name")["unit_supply_price"]
                                    .shift(1).rolling(3).std()
                                    .reset_index(level=0, drop=True))
    else:
        df["price_volatility"] = np.nan

    df["month"] = df["ym"].dt.month
    df["quarter"] = ((df["month"]-1)//3 + 1).astype(int)
    return df

# -----------------------------
# 3) Split 정의
# -----------------------------
train_mask = (monthly_all["ym"] >= "2021-01-01") & (monthly_all["ym"] <= "2023-12-01")
val_mask   = (monthly_all["ym"] >= "2024-01-01") & (monthly_all["ym"] <= "2024-12-01")

monthly_hist = monthly_all[train_mask | val_mask].copy()          # 2021~2024
monthly_hist = build_features(monthly_hist)

# 학습/검증 세트 구성
target = "quantity"
drop_cols = ["ym","product_name",target]
feature_cols = [c for c in monthly_hist.columns if c not in drop_cols]

train_df = monthly_hist[train_mask].copy()
val_df   = monthly_hist[val_mask].copy()

# 결측(초기 라그/롤링) 채움
train_df[feature_cols] = train_df[feature_cols].fillna(0)
val_df[feature_cols]   = val_df[feature_cols].fillna(0)

X_train, y_train = train_df[feature_cols], train_df[target].astype(float)
X_val,   y_val   = val_df[feature_cols],   val_df[target].astype(float)

# -----------------------------
# 4) 모델 후보 정의 (RandomForest, LightGBM, XGBoost)
#     타깃은 log1p로 학습, 평가는 expm1 복원 후 원단위 RMSE/MAE
# -----------------------------
def fit_and_eval(name, model, X_tr, y_tr, X_va, y_va):
    y_tr_log = np.log1p(y_tr)
    model.fit(X_tr, y_tr_log)
    val_pred_log = model.predict(X_va)
    val_pred = np.expm1(val_pred_log)

    rmse = np.sqrt(mean_squared_error(y_va, val_pred))
    mae  = mean_absolute_error(y_va, val_pred)
    return {"name":name, "model":model, "rmse":rmse, "mae":mae}

candidates = [
    ("RandomForest",
     RandomForestRegressor(
         n_estimators=400, max_depth=None, n_jobs=-1, random_state=RANDOM_STATE)),
    ("LightGBM",
     lgb.LGBMRegressor(
         objective="regression", learning_rate=0.05,
         n_estimators=1500, subsample=0.9, colsample_bytree=0.9,
         random_state=RANDOM_STATE)),
]

if HAS_XGB:
    candidates.append(
        ("XGBoost",
         XGBRegressor(
             objective="reg:squarederror", learning_rate=0.05,
             n_estimators=1500, subsample=0.9, colsample_bytree=0.9,
             max_depth=8, tree_method="hist", random_state=RANDOM_STATE))
    )

results = []
for name, model in candidates:
    res = fit_and_eval(name, model, X_train, y_train, X_val, y_val)
    results.append(res)
    print(f"[VAL-{name}] RMSE={res['rmse']:.3f}  MAE={res['mae']:.3f}")

# 베스트 모델 선택 (RMSE 기준)
best = sorted(results, key=lambda x: x["rmse"])[0]
print("\nBest model:", best["name"])

# -----------------------------
# 5) 최종 재학습 (2021~2024 전체로) & 2025-01~03 예측
# -----------------------------
# 5-1) 2021~2024 전체로 재학습
X_tr_full = monthly_hist[feature_cols].fillna(0)
y_tr_full = monthly_hist[target].astype(float)

y_tr_full_log = np.log1p(y_tr_full)
best_model = best["model"]
best_model.fit(X_tr_full, y_tr_full_log)

# 5-2) 2025-Q1 스켈레톤 추가 후 동일 피처 재계산 → 예측
future_months = pd.date_range("2025-01-01","2025-03-01",freq="MS")
products = monthly_all["product_name"].unique()
future_skel = pd.MultiIndex.from_product([products, future_months],
                                         names=["product_name","ym"]).to_frame(index=False)
# 과거(2021~2024) + 미래 스켈레톤 결합
full_for_pred = pd.concat([monthly_all[monthly_all["ym"] <= "2024-12-01"], future_skel],
                          ignore_index=True).sort_values(["product_name","ym"])

# product_id, unit_supply_price 보정
full_for_pred["product_id"] = le.transform(full_for_pred["product_name"])
if "unit_supply_price" in full_for_pred.columns:
    full_for_pred["unit_supply_price"] = (
        full_for_pred.groupby("product_name")["unit_supply_price"].ffill()
    )

# 피처 다시 만들고 2025-01~03만 추출
full_feat = build_features(full_for_pred)
full_feat[feature_cols] = full_feat[feature_cols].fillna(0)
test_df = full_feat[full_feat["ym"].isin(future_months)].copy()

pred_log = best_model.predict(test_df[feature_cols])
test_df["prediction"] = np.expm1(pred_log)

# -----------------------------
# 6) 결과 출력
# -----------------------------
print("\n[Validation summary]")
for r in results:
    print(f"- {r['name']}: RMSE={r['rmse']:.3f}, MAE={r['mae']:.3f}")

print("\n[2025-Q1 prediction: per product (head)]")
print(test_df[["ym","product_name","prediction"]].sort_values(["ym","product_name"]).head(20))

# 월합 비교표 (최근 12개월 + 2025Q1 예측)
actual_agg = (monthly_all[(monthly_all["ym"] >= (monthly_all["ym"].max() - pd.offsets.MonthBegin(11)))]
                .groupby("ym")["quantity"].sum().reset_index(name="actual_quantity"))
pred_agg   = (test_df.groupby("ym")["prediction"].sum().reset_index(name="pred_quantity"))
compare    = pd.concat([actual_agg, pred_agg], axis=0).sort_values("ym")
print("\n[Actual vs Predicted (recent months + 2025Q1)]")
print(compare)
```

### 2. purchase 데이터 모델링

```python
import pandas as pd, numpy as np, holidays

# 날짜 변환
purchase["date"] = pd.to_datetime(purchase["date"], errors="coerce")
purchase["ym"]   = purchase["date"].dt.to_period("M").dt.to_timestamp()

# 단가 생성 (혹시 없다면)
if "unit_price" not in purchase.columns:
    purchase["unit_price"] = (purchase["sales_price"] / purchase["quantity"]).replace([np.inf,-np.inf], np.nan)

# 월별 집계
monthly_p = (purchase.groupby(["product_name","ym"])
             .agg(quantity=("quantity","sum"),
                  unit_price=("unit_price","mean"))
             .reset_index()
             .sort_values(["product_name","ym"]))

# 날짜 기반 파생변수
kr_holidays = holidays.KR(years=range(2021, 2026))
def month_holiday_count(dt):
    days = pd.date_range(dt, dt + pd.offsets.MonthEnd(0), freq="D")
    return int(sum((d.weekday()>=5) or (d in kr_holidays) for d in days))

monthly_p["month"]   = monthly_p["ym"].dt.month
monthly_p["quarter"] = ((monthly_p["month"]-1)//3 + 1).astype(int)
monthly_p["holiday_count"] = monthly_p["ym"].apply(month_holiday_count)
monthly_p["holiday"]       = (monthly_p["holiday_count"] > 0).astype(int)

gq = monthly_p.groupby("product_name")["quantity"]
monthly_p["inbound_qty_lag_1m"]     = gq.shift(1)
monthly_p["inbound_qty_lag_1y"]     = gq.shift(12)
monthly_p["inbound_rolling_avg_3m"] = gq.shift(1).rolling(3).mean().reset_index(level=0, drop=True)
monthly_p["inbound_rolling_max_3m"] = gq.shift(1).rolling(3).max().reset_index(level=0, drop=True)
monthly_p["inbound_rolling_min_3m"] = gq.shift(1).rolling(3).min().reset_index(level=0, drop=True)

monthly_p["price_volatility"] = (monthly_p.groupby("product_name")["unit_price"]
                                 .shift(1).rolling(3).std()
                                 .reset_index(level=0, drop=True))

# Train: 2021~2023, Validation: 2024
train_p = monthly_p.query("'2021-01-01' <= ym <= '2023-12-01'").copy()
val_p   = monthly_p.query("'2024-01-01' <= ym <= '2024-12-01'").copy()

# 예측 대상: 2025-01~03 skeleton 생성
future_months = pd.date_range("2025-01-01","2025-03-01",freq="MS")
prods = monthly_p["product_name"].unique()
future = pd.MultiIndex.from_product([prods, future_months],
                                    names=["product_name","ym"]).to_frame(index=False)
future["month"]   = future["ym"].dt.month
future["quarter"] = ((future["month"]-1)//3 + 1).astype(int)
future["holiday_count"] = future["ym"].apply(month_holiday_count)
future["holiday"]       = (future["holiday_count"] > 0).astype(int)
future["unit_price"]    = np.nan
future["quantity"]      = np.nan

# 과거+미래 합치기
full = (pd.concat([monthly_p, future], ignore_index=True)
          .sort_values(["product_name","ym"]))
full["unit_price"] = full.groupby("product_name")["unit_price"].ffill().fillna(0)

# 다시 lag/rolling 계산 (미래 포함)
gq2 = full.groupby("product_name")["quantity"]
full["inbound_qty_lag_1m"]     = gq2.shift(1)
full["inbound_qty_lag_1y"]     = gq2.shift(12)
full["inbound_rolling_avg_3m"] = gq2.shift(1).rolling(3).mean().reset_index(level=0, drop=True)
full["inbound_rolling_max_3m"] = gq2.shift(1).rolling(3).max().reset_index(level=0, drop=True)
full["inbound_rolling_min_3m"] = gq2.shift(1).rolling(3).min().reset_index(level=0, drop=True)
full["price_volatility"]       = (full.groupby("product_name")["unit_price"]
                                  .shift(1).rolling(3).std()
                                  .reset_index(level=0, drop=True))
import lightgbm as lgb
from sklearn.metrics import mean_squared_error, mean_absolute_error

target = "quantity"
feat_cols = [c for c in full.columns if c not in ["product_name","ym",target,"holiday_count"]]

train_final = full.query("'2021-01-01' <= ym <= '2023-12-01'").copy()
val_final   = full.query("'2024-01-01' <= ym <= '2024-12-01'").copy()
future_25q1 = full[full["ym"].isin(future_months)].copy()

train_final[feat_cols] = train_final[feat_cols].fillna(0)
val_final[feat_cols]   = val_final[feat_cols].fillna(0)
future_25q1[feat_cols] = future_25q1[feat_cols].fillna(0)

# log 변환 타겟
y_train = np.log1p(train_final[target].clip(lower=0))
y_val   = np.log1p(val_final[target].clip(lower=0))

model = lgb.LGBMRegressor(objective="regression", metric="rmse",
                          random_state=42, learning_rate=0.05, n_estimators=800)
model.fit(train_final[feat_cols], y_train,
          eval_set=[(val_final[feat_cols], y_val)],
          eval_metric="rmse",
          callbacks=[lgb.early_stopping(stopping_rounds=50), lgb.log_evaluation(100)])

# Validation 성능
val_pred = np.expm1(model.predict(val_final[feat_cols]))
rmse = np.sqrt(mean_squared_error(val_final[target], val_pred))
mae  = mean_absolute_error(val_final[target], val_pred)
print("Validation RMSE:", rmse)
print("Validation MAE:", mae)

# 2025Q1 예측
future_25q1["prediction"] = np.expm1(model.predict(future_25q1[feat_cols]))

# 상품별 과거 95% 기준 캡핑
hist_p95 = (monthly_p.groupby("product_name")["quantity"]
            .quantile(0.95).rename("p95")).reset_index()
future_25q1 = future_25q1.merge(hist_p95, on="product_name", how="left")
future_25q1["prediction"] = np.minimum(future_25q1["prediction"], future_25q1["p95"])

# 월별 총합
monthly_pred_2025 = future_25q1.groupby("ym")["prediction"].sum().reset_index()
print(monthly_pred_2025)
```

```python
# === Purchase pipeline: 2021~2023 Train / 2024 Val / 2025Q1 Test ===
import pandas as pd, numpy as np, holidays
from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import LabelEncoder
import lightgbm as lgb

# (옵션) XGBoost 사용 가능 여부 확인
try:
    from xgboost import XGBRegressor
    HAS_XGB = True
except Exception:
    HAS_XGB = False

RANDOM_STATE = 42

# -----------------------------
# 0) 기본 정리
# -----------------------------
assert {"date","product_name","quantity"}.issubset(purchase.columns), "필수 컬럼 누락 (date, product_name, quantity)"
df = purchase.copy()
df["date"] = pd.to_datetime(df["date"], errors="coerce")
df = df.dropna(subset=["date"])
df["ym"] = df["date"].dt.to_period("M").dt.to_timestamp()

# 단가 생성(없으면)
if "unit_price" not in df.columns:
    if {"sales_price","quantity"}.issubset(df.columns):
        df["unit_price"] = (df["sales_price"] / df["quantity"]).replace([np.inf,-np.inf], np.nan)
    else:
        df["unit_price"] = np.nan

# -----------------------------
# 1) 월 집계 (입고 기준)
# -----------------------------
monthly_p = (df.groupby(["product_name","ym"])
               .agg(quantity=("quantity","sum"),
                    unit_price=("unit_price","mean"))
               .reset_index()
               .sort_values(["product_name","ym"])
               .reset_index(drop=True))

# 제품 Label Encoding (모든 모델 공통 사용)
le = LabelEncoder()
monthly_p["product_id"] = le.fit_transform(monthly_p["product_name"])

# -----------------------------
# 2) 공휴일/달/분기 + 시계열 피처(미래 누수 방지: 모두 shift(1))
# -----------------------------
kr_holidays = holidays.KR(years=range(2021, 2026))
def month_holiday_count(dt):
    days = pd.date_range(dt, dt + pd.offsets.MonthEnd(0), freq="D")
    return int(sum((d.weekday()>=5) or (d in kr_holidays) for d in days))

def build_features(df):
    df = df.sort_values(["product_name","ym"]).copy()
    df["month"]   = df["ym"].dt.month
    df["quarter"] = ((df["month"]-1)//3 + 1).astype(int)
    df["holiday_count"] = df["ym"].apply(month_holiday_count)
    df["holiday"]       = (df["holiday_count"]>0).astype(int)

    gq = df.groupby("product_name")["quantity"]
    df["inbound_qty_lag_1m"]     = gq.shift(1)
    df["inbound_qty_lag_1y"]     = gq.shift(12)
    df["inbound_rolling_avg_3m"] = gq.shift(1).rolling(3).mean().reset_index(level=0, drop=True)
    df["inbound_rolling_max_3m"] = gq.shift(1).rolling(3).max().reset_index(level=0, drop=True)
    df["inbound_rolling_min_3m"] = gq.shift(1).rolling(3).min().reset_index(level=0, drop=True)

    if "unit_price" in df.columns:
        df["price_volatility"] = (df.groupby("product_name")["unit_price"]
                                    .shift(1).rolling(3).std()
                                    .reset_index(level=0, drop=True))
    else:
        df["price_volatility"] = np.nan
    return df

hist_feat = build_features(monthly_p)

# -----------------------------
# 3) Split: Train=2021~2023, Val=2024
# -----------------------------
train_mask = (hist_feat["ym"] >= "2021-01-01") & (hist_feat["ym"] <= "2023-12-01")
val_mask   = (hist_feat["ym"] >= "2024-01-01") & (hist_feat["ym"] <= "2024-12-01")

target = "quantity"
drop_cols = ["ym","product_name",target]  # holiday_count/holiday 등은 피처로 유지
feat_cols = [c for c in hist_feat.columns if c not in drop_cols]

train_df = hist_feat[train_mask].copy()
val_df   = hist_feat[val_mask].copy()
train_df[feat_cols] = train_df[feat_cols].fillna(0)
val_df[feat_cols]   = val_df[feat_cols].fillna(0)

X_train, y_train = train_df[feat_cols], train_df[target].astype(float)
X_val,   y_val   = val_df[feat_cols],   val_df[target].astype(float)

# -----------------------------
# 4) 모델 3종 비교 (log1p 타깃 학습, expm1로 복원 평가)
# -----------------------------
def fit_and_eval(name, model, X_tr, y_tr, X_va, y_va):
    y_tr_log = np.log1p(y_tr.clip(lower=0))
    model.fit(X_tr, y_tr_log)
    val_pred = np.expm1(model.predict(X_va))
    rmse = np.sqrt(mean_squared_error(y_va, val_pred))
    mae  = mean_absolute_error(y_va, val_pred)
    return {"name":name, "model":model, "rmse":rmse, "mae":mae}

cands = [
    ("RandomForest",
     RandomForestRegressor(n_estimators=500, max_depth=None, n_jobs=-1, random_state=RANDOM_STATE)),
    ("LightGBM",
     lgb.LGBMRegressor(objective="regression", learning_rate=0.05,
                       n_estimators=1500, subsample=0.9, colsample_bytree=0.9,
                       random_state=RANDOM_STATE)),
]
if HAS_XGB:
    cands.append(
        ("XGBoost",
         XGBRegressor(objective="reg:squarederror", learning_rate=0.05,
                      n_estimators=1500, subsample=0.9, colsample_bytree=0.9,
                      max_depth=8, tree_method="hist", random_state=RANDOM_STATE))
    )

results = []
for name, mdl in cands:
    res = fit_and_eval(name, mdl, X_train, y_train, X_val, y_val)
    results.append(res)
    print(f"[VAL-{name}] RMSE={res['rmse']:.3f}  MAE={res['mae']:.3f}")

best = sorted(results, key=lambda x: x["rmse"])[0]
print("\nBest model:", best["name"])

# -----------------------------
# 5) 최종 재학습(2021~2024 전체) & 2025-01~03 예측
# -----------------------------
# 5-1) 재학습
X_full = hist_feat[feat_cols].fillna(0)  # 2021~2024
y_full = hist_feat[target].astype(float)
best_model = best["model"]
best_model.fit(X_full, np.log1p(y_full.clip(lower=0)))

# 5-2) 2025-Q1 스켈레톤 결합 후 동일 피처 재계산
future_months = pd.date_range("2025-01-01","2025-03-01",freq="MS")
prods = monthly_p["product_name"].unique()
future = pd.MultiIndex.from_product([prods, future_months],
                                    names=["product_name","ym"]).to_frame(index=False)
# 과거 + 미래 결합 (단가 ffill)
full = (pd.concat([monthly_p, future], ignore_index=True)
          .sort_values(["product_name","ym"]).reset_index(drop=True))
full["product_id"] = le.transform(full["product_name"])
full["unit_price"] = full.groupby("product_name")["unit_price"].ffill()

full_feat = build_features(full)
full_feat[feat_cols] = full_feat[feat_cols].fillna(0)
test_df = full_feat[full_feat["ym"].isin(future_months)].copy()

pred_lin = np.expm1(best_model.predict(test_df[feat_cols]))
test_df["prediction"] = np.clip(pred_lin, a_min=0, a_max=None)

# -----------------------------
# 6) 안전 캡핑(제품별 과거 95% 상한) & 월합 출력
# -----------------------------
hist_p95 = (monthly_p.groupby("product_name")["quantity"]
            .quantile(0.95).rename("p95")).reset_index()
test_df = test_df.merge(hist_p95, on="product_name", how="left")
test_df["prediction"] = np.minimum(test_df["prediction"], test_df["p95"])

print("\n[Validation summary]")
for r in results:
    print(f"- {r['name']}: RMSE={r['rmse']:.3f}, MAE={r['mae']:.3f}")

print("\n[2025-Q1 prediction: per product (head)]")
print(test_df[["ym","product_name","prediction"]].sort_values(["ym","product_name"]).head(20))

monthly_pred_2025 = test_df.groupby("ym")["prediction"].sum().reset_index()
print("\n[2025-Q1 monthly total inbound prediction]")
print(monthly_pred_2025)
```

```python
##=============================================================================
##1.purchase&Sales 데이터 불러오기=============================================-
##=============================================================================
purchase=pd.read_csv('B_purchase_1.csv')
purchase['date']=pd.to_datetime(purchase['date'])

# quantity 또는 supply_price 가 음수이거나 0인 행 제거
df_cleaned = purchase[(purchase["quantity"] >= 0) & (purchase["sales_price"] >= 0)].copy()

#판매수량이나 가격이 음수이인 행 제거
purchase=purchase[(purchase["quantity"] >= 0) & (purchase["sales_price"] >= 0)]

##==========quarter 변수 생성==================

# month 기준으로 quarter 변수 생성
purchase["quarter"] = ((purchase["month"] - 1) // 3) + 1

######holiday: 주말 또는 공휴일이면 1, 아니면 0#######
import holidays

# 날짜 범위 생성
date_range = pd.date_range(start="2021-01-01", end="2025-03-31", freq="D")
df = pd.DataFrame({"date": date_range})

# 한국 공휴일 세팅
kr_holidays = holidays.KR(years=[2021, 2022, 2023, 2024, 2025])

# 공휴일 여부
df["is_holiday"] = df["date"].isin(kr_holidays)

# 주말 여부 (토:5, 일:6)
df["is_weekend"] = df["date"].dt.weekday >= 5

#  holiday 변수 생성 (공휴일 or 주말 → 1, 그 외 → 0)
df["holiday"] = ((df["is_holiday"]) | (df["is_weekend"])).astype(int)
df = df[["date", "holiday"]]

# holiday df와 원본 데이터 merge
purchase=purchase.merge(df[["date", "holiday"]], on="date", how="left")

# 칼럼 삭제
purchase.drop(['client_id', 'client_zipcode',
       'supplier_zipcode', 'stock_in_type', 'barcode', 'spec',
        'option_code', 'option_name', 'units_per_pack',
        'each_count','vat', 'main_category',
       'mid_category', 'sub_category'], axis=1, inplace=True)

##========Sales 데이터 불러오기==========================
sales=pd.read_csv('B_sales_1.csv')
sales['sales_date']=pd.to_datetime(sales['sales_date'])

# quantity 또는 supply_price 가 음수이거나 0인 행 제거
df_cleaned = sales[(sales["quantity"] >= 0) & (sales["supply_price"] >= 0)].copy()


#판매수량이나 가격이 음수이인 행 제거
sales=sales[(sales["quantity"] >= 0) & (sales["supply_price"] >= 0)]

##### month 기준으로 quarter 변수 생성######
sales["quarter"] = ((sales["month"] - 1) // 3) + 1

#칼럼 삭제
sales.drop(['spec','units_per_pack','client_zipcode',
            'client_id','option_code','main_category','mid_category',
            'sub_category','vat'], axis=1, inplace=True)

##=============================================================================
##============ purchase 데이터 기반 파생변수 생성==============================
##=============================================================================

import re
import pandas as pd
import numpy as np

# Step 1) 탄산음료 파생변수

def split_name(name: str):
    maker = re.findall(r"\((.*?)\)", str(name))
    maker = maker[0] if maker else None
    cleaned = re.sub(r"\(.*?\)", "", str(name))
    product = re.split(r"\d", cleaned)[0]
    return maker, product.strip()

# 제조사(company), 제로 여부
purchase[["company", "name"]] = purchase["product_name"].apply(lambda x: pd.Series(split_name(x)))

# Step 2) name × 연월 집계

PURCHASE_DATE_COL = "date"
QTY_COL           = "quantity"
PRICE_COL         = "sales_price"
GROUP_KEY_COLS    = ["product_name"]

def _ensure_datetime(df, date_col):
    df = df.copy()
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
    return df

def _make_key(df, key_cols):
    if len(key_cols) == 1:
        return df[key_cols[0]].astype(str)
    return df[key_cols].astype(str).agg("§".join, axis=1)

# 기본 전처리
purchase = _ensure_datetime(purchase, PURCHASE_DATE_COL)
purchase["_grp_key"] = _make_key(purchase, GROUP_KEY_COLS)
purchase["_ym"] = purchase[PURCHASE_DATE_COL].dt.to_period("M").astype(str)

# 연월 기준 집계
monthly = (
    purchase.groupby(["_grp_key", "_ym"], as_index=False)
    .agg({
        QTY_COL   : "sum",   # 월별 총 수량
        PRICE_COL : "sum",   # 월별 총 금액
        "holiday" : "sum",   # 월별 주말/공휴일 주문 건수
    })
    .rename(columns={
        QTY_COL: "quantity",            # inbound_qty → quantity
        PRICE_COL: "sales_price",       # 매입금액 합계
        "holiday": "holiday_count",
    })
)

# Step 3) 시간 변수 & 정렬

monthly["_ym_period"] = pd.to_datetime(monthly["_ym"] + "-01")
monthly = monthly.sort_values(["_grp_key", "_ym_period"])
monthly["year"] = monthly["_ym"].str[:4].astype(int)
monthly["month"] = monthly["_ym"].str[5:7].astype(int)
monthly["quarter"] = ((monthly["month"] - 1) // 3) + 1

# Step 4) lag & rolling

def _lag_roll(g, window=3):
    g = g.copy()
    g["inbound_qty_lag_1m"] = g["quantity"].shift(1)
    g["inbound_qty_lag_1y"] = g["quantity"].shift(12)
    g["inbound_rolling_avg_3m"] = g["quantity"].rolling(window).mean()
    g["inbound_rolling_max_3m"] = g["quantity"].rolling(window).max()
    g["inbound_rolling_min_3m"] = g["quantity"].rolling(window).min()
    return g
monthly = monthly.groupby("_grp_key", group_keys=False).apply(_lag_roll, window=3)


# Step 5) 단가 & 가격 변동성

monthly["unit_price"] = monthly["sales_price"] / monthly["quantity"].replace(0, np.nan)
def _price_vol(g, window=3):
    g = g.copy()
    g["price_volatility"] = g["unit_price"].rolling(window=window, min_periods=2).std()
    return g
monthly = monthly.groupby("_grp_key", group_keys=False).apply(_price_vol, window=3)


# Step 6) 최종 정리

monthly = monthly.drop(columns=["_ym_period"]).rename(columns={"_grp_key": "product_name"})
monthly = monthly.sort_values(["year", "month"]).reset_index(drop=True)
monthly = monthly[
    ["product_name", "year", "month", "quarter",
     "quantity", "sales_price", "holiday_count",
     "inbound_qty_lag_1m", "inbound_qty_lag_1y",
     "inbound_rolling_avg_3m", "inbound_rolling_max_3m", "inbound_rolling_min_3m",
     "unit_price", "price_volatility"]
]


# ===========탄산 음료만의 추가 파생변수=========
# 제로 유무 추가
monthly["is_zero"] = monthly["product_name"].str.contains("제로", na=False).astype(int)

# 음료 용량에 관해서
def extract_volume(name):
    matches = re.findall(r"(\d+[.,]?\d*)(ml|ML|l|L|ℓ)?", name)
    if matches:
        val, unit = matches[-1]  # 마지막 매칭
        val = float(val.replace(",", "."))  # 1,5 → 1.5

        if not unit:
            val *= 1000
        elif unit.lower() in ["l", "ℓ"]:
            val *= 1000

        return val
    return np.nan

monthly["volume_ml"] = monthly["product_name"].apply(extract_volume)
monthly_purchase=monthly.copy()

##=============================================================================
##============ sales 데이터 기반 파생변수 생성================================
##=============================================================================

# Step 1) 탄산음료 파생변수

import re
import pandas as pd
import numpy as np

def split_name(name: str):
    maker = re.findall(r"\((.*?)\)", str(name))
    maker = maker[0] if maker else None
    cleaned = re.sub(r"\(.*?\)", "", str(name))
    product = re.split(r"\d", cleaned)[0]
    return maker, product.strip()

# 제조사(company), 제로 여부
sales[["company", "name"]] = sales["product_name"].apply(lambda x: pd.Series(split_name(x)))

# Step 2) name × 연월 집계
SALES_DATE_COL = "sales_date"
QTY_COL           = "quantity"
PRICE_COL         = "supply_price"
GROUP_KEY_COLS    = ["product_name"]

def _ensure_datetime(df, date_col):
    df = df.copy()
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
    return df

def _make_key(df, key_cols):
    if len(key_cols) == 1:
        return df[key_cols[0]].astype(str)
    return df[key_cols].astype(str).agg("§".join, axis=1)

# 기본 전처리
sales = _ensure_datetime(sales, SALES_DATE_COL)
sales["_grp_key"] = _make_key(sales, GROUP_KEY_COLS)
sales["_ym"] = sales[SALES_DATE_COL].dt.to_period("M").astype(str)

# 연월 기준 집계
monthly_sales = (
    sales.groupby(["_grp_key", "_ym"], as_index=False)
    .agg({
        QTY_COL   : "sum",   # 월별 총 수량
        PRICE_COL : "sum",   # 월별 총 금액
    })
    .rename(columns={
        QTY_COL: "quantity",            # inbound_qty → quantity
        PRICE_COL: "supply_price",       # 매입금액 합계
    })
)

# Step 3) 시간 변수 & 정렬
monthly_sales["_ym_period"] = pd.to_datetime(monthly_sales["_ym"] + "-01")
monthly_sales = monthly_sales.sort_values(["_grp_key", "_ym_period"])

monthly_sales["year"] = monthly_sales["_ym"].str[:4].astype(int)
monthly_sales["month"] = monthly_sales["_ym"].str[5:7].astype(int)
monthly_sales["quarter"] = ((monthly_sales["month"] - 1) // 3) + 1

# Step 4) lag & rolling
def _sales_lag_roll(g, window=3):
    g = g.copy()
    g["sales_qty_lag_1m"] = g["quantity"].shift(1)
    g["sales_qty_lag_1y"] = g["quantity"].shift(12)

    g["sales_rolling_avg_3m"] = g["quantity"].rolling(window).mean()
    g["sales_rolling_max_3m"] = g["quantity"].rolling(window).max()
    g["sales_rolling_min_3m"] = g["quantity"].rolling(window).min()
    return g

monthly_sales = monthly_sales.groupby("_grp_key", group_keys=False).apply(_sales_lag_roll, window=3) # Corrected function name

# Step 5) 단가 & 가격 변동성
monthly_sales["unit_supply_price"] = monthly_sales["supply_price"] / monthly_sales["quantity"].replace(0, np.nan)

def _price_vol(g, window=3):
    g = g.copy()
    g["price_volatility"] = g["unit_supply_price"].rolling(window=window, min_periods=2).std()
    return g
monthly_sales = monthly_sales.groupby("_grp_key", group_keys=False).apply(_price_vol, window=3)

# Step 6) 최종 정리

monthly_sales = monthly_sales.drop(columns=["_ym_period"]).rename(columns={"_grp_key": "product_name"})
monthly_sales = monthly_sales.sort_values(["year", "month"]).reset_index(drop=True)
monthly_sales = monthly_sales[
    ["product_name", "year", "month", "quarter",
     "quantity", "supply_price",
     "sales_qty_lag_1m", "sales_qty_lag_1y",
     "sales_rolling_avg_3m", "sales_rolling_max_3m", "sales_rolling_min_3m",
     "unit_supply_price", "price_volatility"]
]

# ===========탄산 음료만의 추가 파생변수=========
# 제로 유무 추가
monthly_sales["is_zero"] = monthly_sales["product_name"].str.contains("제로", na=False).astype(int)

# 음료 용량에 관해서
def extract_volume(name):
    matches = re.findall(r"(\d+[.,]?\d*)(ml|ML|l|L|ℓ)?", name)
    if matches:
        val, unit = matches[-1]  # 마지막 매칭
        val = float(val.replace(",", "."))  # 1,5 → 1.5
        if not unit:
            val *= 1000
        elif unit.lower() in ["l", "ℓ"]:
            val *= 1000
        return val
    return np.nan

monthly_sales["volume_ml"] = monthly_sales["product_name"].apply(extract_volume)

##=============================================================================
##============ 공공데이터를 이용한 파생변수 생성================================
##=============================================================================

#=======날씨 관련 변수 생성========
weather=pd.read_csv('weather_data.csv')

#=======경제 관련 변수 생성=======
cpi=pd.read_csv('소비자물가지수.csv')
survey=pd.read_csv('소비자심리지수.csv', encoding='cp949')
survey=survey[(survey['CSI코드별']=='소비자심리지수')&(survey['CSI분류코드별']=='전체')]
cpi_soda=pd.read_excel('소비자물가지수_탄산음료.xlsx')
cpi_soda.drop(['지출목적별'], axis=1, inplace=True)

# 소비자심리지수 처리 :ccsi
csi_clean = survey.copy()
csi_clean['시점'] = csi_clean['시점'].str.replace(" 월","",regex=False)
csi_clean[['year','month']] = csi_clean['시점'].str.split('.',expand=True).astype(int)
csi_clean = csi_clean[['year','month','소비자동향조사(전국, 월, 2008.9~)']]
csi_clean.rename(columns={'소비자동향조사(전국, 월, 2008.9~)':'ccsi'}, inplace=True)

# 소비자물가지수 처리 (wide → long 변환) :cpi
cpi_clean = cpi[cpi['시도별']=='전국'].drop(columns=['시도별'])
cpi_long = cpi_clean.melt(var_name='date', value_name='cpi')
cpi_long[['year','month']] = cpi_long['date'].str.split('.',expand=True).astype(int)
cpi_long = cpi_long[['year','month','cpi']]


#소비자 물가지수(탄산음료) :cpi_soda
cpi_soda_clean = cpi_soda[cpi_soda['시도별']=='전국'].drop(columns=['시도별'])
cpi_soda_long = cpi_soda_clean.melt(var_name='date', value_name='cpi_soda')
cpi_soda_long[['year','month']] = cpi_soda_long['date'].str.split('.',expand=True).astype(int)
cpi_soda_long = cpi_soda_long[['year','month','cpi_soda']]

# 병합
merged = pd.merge(csi_clean, cpi_long, on=["year","month"], how="outer")
merged = pd.merge(merged, cpi_soda_long, on=["year","month"], how="outer")
merged = pd.merge(merged, weather, on=["year","month"], how="outer")
merged = merged.sort_values(["year","month"]).reset_index(drop=True)

# ===========구글 트렌드 변수(탄산음료)================

# 파일 불러오기 (앞 2줄 메타데이터 제거)
google = pd.read_csv("multiTimeline (1).csv", skiprows=2)

# 날짜 컬럼 확인 후 변환 (보통 '주' 혹은 'Week' 라벨로 되어 있음)
google.rename(columns={google.columns[0]: "date"}, inplace=True)
google["date"] = pd.to_datetime(google["date"], errors="coerce")

# 연, 월 추출
google["year"] = google["date"].dt.year
google["month"] = google["date"].dt.month

# 월별 합계 계산
google_avg = (
    google.groupby(["year", "month"], as_index=False)
      .sum(numeric_only=True)
)
google_avg.columns=["year","month","trend_soda_google"]
google_avg= google_avg.sort_values(["year","month"]).reset_index(drop=True)
google_avg=google_avg.iloc[1:,:]

# ========== 네이버 트렌드 변수(탄산음료)================
# 필요한 부분만 추출
naver=pd.read_excel('trend_soda_naver.xlsx')
naver_clean = naver.copy()
naver_clean.columns = ["date", "탄산음료"]

# 날짜 변환
naver_clean["date"] = pd.to_datetime(naver_clean["date"], errors="coerce")
naver_clean["year"] = naver_clean["date"].dt.year
naver_clean["month"] = naver_clean["date"].dt.month
naver_clean.drop(['date'], axis=1, inplace=True)
naver_clean.columns=["trend_soda_naver","year","month"]
naver_clean= naver_clean.sort_values(["year","month"]).reset_index(drop=True)

#트랜드 변수 관련 데이터 병합
merged_trend = pd.merge(google_avg, naver_clean, on=["year","month"], how="outer")
merged_trend= merged_trend.sort_values(["year","month"]).reset_index(drop=True)
merged_trend

#공공데이터 데이터 총 병합
merged_fin =merged.merge(merged_trend, on=["year","month"], how="left")

#최종 병합
sales_fin = monthly_sales.merge(merged_fin, on=["year","month"], how="left")
purchase_fin =monthly_purchase.merge(merged_fin, on=["year","month"], how="left")

##=============================================================================
##============ 2025년 1월-3월 데이터 생성하기(레퍼런스 만들기)=================
##=============================================================================
def make_q4_ref_and_future(
    data: pd.DataFrame,
    key_col="product_name",
    year_col="year",
    month_col="month",
    target_cols=("quantity",),
):
    df = data.copy()

    #  레퍼런스 만들기
    df["yyyymm"] = df[year_col].astype(int) * 100 + df[month_col].astype(int)
    q4 = df[(df[year_col] == 2024) & (df[month_col].isin([10, 11, 12]))].copy()
    q4 = q4.sort_values([key_col, "yyyymm"])
    time_cols = [c for c in [year_col, month_col, "day", "yyyymm"] if c in df.columns]
    exclude_cols = set([key_col] + time_cols + list(target_cols))
    cols_to_uniform = [c for c in df.columns if c not in exclude_cols]

    # Q4 최신 row 추출
    q4_latest = (
        q4.sort_values([key_col, "yyyymm"])
          .groupby(key_col)
          .last()
          .reset_index()
    )
    # product-level 속성만 ref에 포함
    ref = q4_latest[[key_col] + cols_to_uniform]

    # 전체 데이터에 레퍼런스 머지
    merged = df.merge(ref, on=key_col, how="left", suffixes=("", "_q4ref"))
    for c in cols_to_uniform:
        merged[c] = merged[f"{c}_q4ref"].combine_first(merged[c])
        merged.drop(columns=[f"{c}_q4ref"], inplace=True)
    merged.drop(columns=["yyyymm"], errors="ignore", inplace=True)
    df_uniform = merged

    # 2025.01~03 skeleton 생성
    items_2025 = ref[key_col].drop_duplicates()
    future_idx = pd.MultiIndex.from_product(
        [[2025], [1, 2, 3], items_2025],
        names=[year_col, month_col, key_col]
    )
    future = future_idx.to_frame(index=False)
    future = future.merge(ref, on=key_col, how="left")

    # quarter 자동 생성
    future["quarter"] = ((future[month_col] - 1) // 3 + 1)

    # 예측 대상(y) 초기화
    for c in target_cols:
        if c in df.columns:
            future[c] = np.nan
    if "day" in df.columns:
        future["day"] = 1
    return df_uniform, future

df_purchase, future_purchase = make_q4_ref_and_future(purchase_fin, key_col="product_name")
df_sales, future_sales = make_q4_ref_and_future(sales_fin, key_col="product_name")

future_purchase.drop(['ccsi', 'cpi', 'cpi_soda',
       'temperature', 'rain', 'trend_soda_google', 'trend_soda_naver'], axis=1, inplace=True)

future_sales.drop(['ccsi', 'cpi', 'cpi_soda',
       'temperature', 'rain', 'trend_soda_google', 'trend_soda_naver'], axis=1, inplace=True)

future_purchase=future_purchase.merge(merged_fin, on=["year","month"], how="left")
future_sales=future_sales.merge(merged_fin, on=["year","month"], how="left")

purchase_extended=pd.concat([purchase_fin,future_purchase])
sales_extended=pd.concat([sales_fin,future_sales])
purchase_extended.drop('holiday_count', axis=1, inplace=True)
```

```python
import pandas as pd
import holidays
import numpy as np

df1 = pd.read_csv('B_sales_2.csv')
df2 = pd.read_csv('B_purchase_2.csv')
df3 = pd.read_csv('weather_data.csv')
df4 = pd.read_csv('품목별_소비자물가지수.csv', encoding='cp949')
df5 = pd.read_csv('소비자물가지수.csv')
df6 = pd.read_csv('소비자심리지수.csv', encoding='cp949')
df7 = pd.read_csv('B_sales.csv')
df8 = pd.read_csv('B_purchase.csv')
df9 = pd.read_excel('네이버_데이터랩_봉지라면.xlsx')
df10 = pd.read_excel('네이버_데이터랩_컵라면.xlsx')

sales = df1.copy()
purchase = df2.copy()
weather = df3.copy()
cpi_ramyeon = df4.copy()
cpi = df5.copy()
ccsi = df6.copy()
B_sales = df7.copy()
B_purchase = df8.copy()
trends_ramyeon = df9.copy()
trends_cup_noodles = df10.copy()

# ============================================================================
# 1. 모든 외부 데이터 전처리
# ============================================================================

# --- 1-1. 품목별(라면) 소비자물가지수 (cpi_ramyeon) ---
cpi_ramyeon['시점'] = pd.to_datetime(cpi_ramyeon['시점'], format='%Y.%m 월')
cpi_ramyeon = cpi_ramyeon[cpi_ramyeon['품목별'] == '라면'].iloc[:, [2,3]]
cpi_ramyeon.columns = ['date', 'cpi_ramyeon']
cpi_ramyeon['year'] = cpi_ramyeon['date'].dt.year
cpi_ramyeon['month'] = cpi_ramyeon['date'].dt.month
cpi_ramyeon.drop(columns=['date'], inplace=True)

# --- 1-2. 전체 소비자물가지수 (cpi) ---
cpi = pd.melt(cpi, id_vars=['시도별'], var_name='date', value_name='cpi')
cpi['date'] = pd.to_datetime(cpi['date'], format='%Y.%m')
cpi['year'] = cpi['date'].dt.year
cpi['month'] = cpi['date'].dt.month
cpi = cpi.drop(columns=['시도별', 'date']).drop_duplicates()

# --- 1-3. 소비자심리지수 (ccsi) ---
ccsi['시점'] = pd.to_datetime(ccsi['시점'], format='%Y.%m 월')
ccsi = ccsi[ccsi['CSI코드별'] == '소비자심리지수'].iloc[:,[2, 3]]
ccsi.columns = ['date', 'ccsi']
ccsi['year'] = ccsi['date'].dt.year
ccsi['month'] = ccsi['date'].dt.month
ccsi.drop(columns=['date'], inplace=True)

# --- 1-4. '용기라면' 과거 판매량 변수 ---
B_sales['sales_date'] = pd.to_datetime(B_sales['sales_date'])
cup_noodle_sales = B_sales[B_sales['sub_category'] == '용기라면'].copy()
cup_noodle_monthly_sales = cup_noodle_sales.set_index('sales_date').resample('MS')['quantity'].sum().reset_index()
# 고유한 이름으로 변경
cup_noodle_monthly_sales.rename(columns={'quantity': 'cup_noodle_sales_qty'}, inplace=True)
cup_noodle_monthly_sales['cup_noodle_sales_lag_1m'] = cup_noodle_monthly_sales['cup_noodle_sales_qty'].shift(1)
cup_noodle_monthly_sales['cup_noodle_sales_rolling_avg_3m'] = cup_noodle_monthly_sales['cup_noodle_sales_qty'].rolling(3, min_periods=1).mean()
cup_noodle_monthly_sales['year'] = cup_noodle_monthly_sales['sales_date'].dt.year
cup_noodle_monthly_sales['month'] = cup_noodle_monthly_sales['sales_date'].dt.month

B_purchase['date'] = pd.to_datetime(B_purchase['date'])
cup_noodle_purchase = B_purchase[B_purchase['sub_category'] == '용기라면'].copy()
cup_noodle_monthly_purchase = cup_noodle_purchase.set_index('date').resample('MS')['quantity'].sum().reset_index()
# 고유한 이름으로 변경
cup_noodle_monthly_purchase.rename(columns={'quantity': 'cup_noodle_purchase_qty'}, inplace=True)
cup_noodle_monthly_purchase['cup_noodle_purchase_lag_1m'] = cup_noodle_monthly_purchase['cup_noodle_purchase_qty'].shift(1)
cup_noodle_monthly_purchase['cup_noodle_purchase_rolling_avg_3m'] = cup_noodle_monthly_purchase['cup_noodle_purchase_qty'].rolling(3, min_periods=1).mean()
cup_noodle_monthly_purchase['year'] = cup_noodle_monthly_purchase['date'].dt.year
cup_noodle_monthly_purchase['month'] = cup_noodle_monthly_purchase['date'].dt.month

# --- 1-5. 온라인 트렌드 데이터 (네이버 데이터랩) ---
trends_ramyeon = trends_ramyeon.iloc[6:].reset_index(drop=True)
trends_ramyeon.columns = ['url', 'trends_ramyeon']
trends_ramyeon['url'] = pd.to_datetime(trends_ramyeon['url'])
trends_ramyeon['year'] = trends_ramyeon['url'].dt.year
trends_ramyeon['month'] = trends_ramyeon['url'].dt.month
trends_ramyeon.drop(columns=['url'], inplace=True)

trends_cup_noodles = trends_cup_noodles.iloc[6:].reset_index(drop=True)
trends_cup_noodles.columns = ['url', 'trends_cup_noodles']
trends_cup_noodles['url'] = pd.to_datetime(trends_cup_noodles['url'])
trends_cup_noodles['year'] = trends_cup_noodles['url'].dt.year
trends_cup_noodles['month'] = trends_cup_noodles['url'].dt.month
trends_cup_noodles.drop(columns=['url'], inplace=True)

# --- 1-6. 공휴일 데이터 ---
start_date = min(sales['sales_date'].min(), purchase['date'].min())
end_date = max(sales['sales_date'].max(), purchase['date'].max())
calendar = pd.DataFrame(pd.date_range(start=start_date, end=end_date, freq='D'), columns=['date_col'])
all_years = calendar['date_col'].dt.year.unique()
kr_holidays = holidays.KR(years=all_years)
kr_holidays_list = pd.to_datetime(list(kr_holidays.keys()))
calendar['holiday'] = ((calendar['date_col'].dt.dayofweek >= 5) | (calendar['date_col'].isin(kr_holidays_list))).astype(int)
monthly_holiday_count = calendar.resample('MS', on='date_col')['holiday'].sum().reset_index()
monthly_holiday_count.rename(columns={'holiday': 'holiday_count', 'date_col': 'date'}, inplace=True)
monthly_holiday_count['year'] = monthly_holiday_count['date'].dt.year
monthly_holiday_count['month'] = monthly_holiday_count['date'].dt.month

# ============================================================================
# 2. 과거 데이터 월별 집계 및 모든 변수 생성 (2024년 12월까지)
# ============================================================================

# --- 2-1. 월별 집계 및 기본 변수 생성 ---
sales['sales_date'] = pd.to_datetime(sales['sales_date'])
purchase['date'] = pd.to_datetime(purchase['date'])
sales_agg_dict = {'quantity': 'sum', 'supply_price': 'sum'}
purchase_agg_dict = {'quantity': 'sum', 'sales_price': 'sum'}
monthly_sales_hist = sales.groupby('product_name').resample('MS', on='sales_date').agg(sales_agg_dict).reset_index()
monthly_purchase_hist = purchase.groupby('product_name').resample('MS', on='date').agg(purchase_agg_dict).reset_index()

for df, date_col in zip([monthly_sales_hist, monthly_purchase_hist], ['sales_date', 'date']):
    df['year'] = df[date_col].dt.year
    df['month'] = df[date_col].dt.month
    df['quarter'] = df[date_col].dt.quarter

# --- 2-2. 기술 통계 및 가격 변수 생성 ---
monthly_sales_hist['sales_qty_lag_1m'] = monthly_sales_hist.groupby('product_name')['quantity'].shift(1)
monthly_sales_hist['sales_qty_lag_1y'] = monthly_sales_hist.groupby('product_name')['quantity'].shift(12)
monthly_purchase_hist['inbound_qty_lag_1m'] = monthly_purchase_hist.groupby('product_name')['quantity'].shift(1)
monthly_purchase_hist['inbound_qty_lag_1y'] = monthly_purchase_hist.groupby('product_name')['quantity'].shift(12)

monthly_sales_hist['sales_rolling_avg_3m'] = monthly_sales_hist.groupby('product_name')['quantity'].rolling(3, min_periods=1).mean().reset_index(level=0, drop=True)
monthly_sales_hist['sales_rolling_max_3m'] = monthly_sales_hist.groupby('product_name')['quantity'].rolling(3, min_periods=1).max().reset_index(level=0, drop=True)
monthly_sales_hist['sales_rolling_min_3m'] = monthly_sales_hist.groupby('product_name')['quantity'].rolling(3, min_periods=1).min().reset_index(level=0, drop=True)
monthly_purchase_hist['inbound_rolling_avg_3m'] = monthly_purchase_hist.groupby('product_name')['quantity'].rolling(3, min_periods=1).mean().reset_index(level=0, drop=True)
monthly_purchase_hist['inbound_rolling_max_3m'] = monthly_purchase_hist.groupby('product_name')['quantity'].rolling(3, min_periods=1).max().reset_index(level=0, drop=True)
monthly_purchase_hist['inbound_rolling_min_3m'] = monthly_purchase_hist.groupby('product_name')['quantity'].rolling(3, min_periods=1).min().reset_index(level=0, drop=True)

product_category_map = sales[['product_name', 'sub_category']].drop_duplicates()
monthly_sales_hist = pd.merge(monthly_sales_hist, product_category_map, on='product_name', how='left')
monthly_purchase_hist = pd.merge(monthly_purchase_hist, product_category_map, on='product_name', how='left')

monthly_sales_hist['unit_supply_price'] = monthly_sales_hist['supply_price'] / monthly_sales_hist['quantity']
monthly_purchase_hist['unit_price'] = monthly_purchase_hist['sales_price'] / monthly_purchase_hist['quantity']
monthly_sales_hist['category_avg_price'] = monthly_sales_hist.groupby([monthly_sales_hist['sales_date'], 'sub_category'])['unit_supply_price'].transform('mean')
monthly_sales_hist['price_vs_category_avg'] = monthly_sales_hist['unit_supply_price'] / monthly_sales_hist['category_avg_price']
monthly_purchase_hist['category_avg_price'] = monthly_purchase_hist.groupby([monthly_purchase_hist['date'], 'sub_category'])['unit_price'].transform('mean')
monthly_purchase_hist['price_vs_category_avg'] = monthly_purchase_hist['unit_price'] / monthly_purchase_hist['category_avg_price']
monthly_sales_hist['price_volatility'] = monthly_sales_hist.groupby('product_name')['unit_supply_price'].rolling(3, min_periods=1).std().reset_index(level=0, drop=True)
monthly_purchase_hist['price_volatility'] = monthly_purchase_hist.groupby('product_name')['unit_price'].rolling(3, min_periods=1).std().reset_index(level=0, drop=True)

# --- [수정] 2-3. 과거 데이터에 모든 외부 정보 병합 ---
common_external_data = [monthly_holiday_count.drop(columns='date'), weather, cpi_ramyeon, cpi, ccsi, trends_ramyeon, trends_cup_noodles]
for df in common_external_data:
    monthly_sales_hist = pd.merge(monthly_sales_hist, df, on=['year', 'month'], how='left')
    monthly_purchase_hist = pd.merge(monthly_purchase_hist, df, on=['year', 'month'], how='left')

# '용기라면' 데이터를 이름이 바뀌었으므로 각각 명시적으로 병합
monthly_sales_hist = pd.merge(monthly_sales_hist, cup_noodle_monthly_sales.drop(columns='sales_date'), on=['year', 'month'], how='left')
monthly_purchase_hist = pd.merge(monthly_purchase_hist, cup_noodle_monthly_purchase.drop(columns='date'), on=['year', 'month'], how='left')

# --- 2-4. 과거 데이터 최종 정리 ---
monthly_sales_hist.drop(columns=['category_avg_price'], inplace=True, errors='ignore')
monthly_purchase_hist.drop(columns=['category_avg_price'], inplace=True, errors='ignore')
monthly_sales_hist.replace([np.inf, -np.inf], 0, inplace=True)
monthly_purchase_hist.replace([np.inf, -np.inf], 0, inplace=True)
monthly_sales_hist.fillna(0, inplace=True)
monthly_purchase_hist.fillna(0, inplace=True)

# ============================================================================
# 3. 2025년 미래 데이터 생성을 위한 '정답지(Lookup Tables)' 준비
# ============================================================================
# Sales용 정답지
sales_last_month = monthly_sales_hist.sort_values(by='sales_date').drop_duplicates(subset='product_name', keep='last')
sales_q4_stats = monthly_sales_hist[monthly_sales_hist['year'] == 2024].groupby('product_name')['quantity'].agg(['mean', 'max', 'min']).reset_index()

# --- [수정] '용기라면' 관련 변수들을 다시 정답지에 포함시킵니다. ---
sales_lookup = sales_last_month[['product_name', 'quantity', 'unit_supply_price', 'price_vs_category_avg',
                                 'cup_noodle_sales_qty', 'cup_noodle_sales_lag_1m', 'cup_noodle_sales_rolling_avg_3m']].copy()

sales_lookup.rename(columns={'quantity': 'sales_qty_lag_1m_lookup', 'unit_supply_price': 'unit_supply_price_lookup', 'price_vs_category_avg': 'pvc_lookup'}, inplace=True)
sales_lookup = pd.merge(sales_lookup, sales_q4_stats.rename(columns={'mean': 'sra_lookup', 'max': 'srmax_lookup', 'min': 'srmin_lookup'}), on='product_name', how='left')
sales_2024_early_lookup = monthly_sales_hist[monthly_sales_hist['year'] == 2024][['product_name', 'month', 'quantity']].rename(columns={'quantity': 'lag_1y_lookup'})

# Purchase용 정답지
purchase_last_month = monthly_purchase_hist.sort_values(by='date').drop_duplicates(subset='product_name', keep='last')
purchase_q4_stats = monthly_purchase_hist[monthly_purchase_hist['year'] == 2024].groupby('product_name')['quantity'].agg(['mean', 'max', 'min']).reset_index()

# --- [수정] '용기라면' 관련 변수들을 다시 정답지에 포함시킵니다. ---
purchase_lookup = purchase_last_month[['product_name', 'quantity', 'unit_price', 'price_vs_category_avg',
                                       'cup_noodle_purchase_qty', 'cup_noodle_purchase_lag_1m', 'cup_noodle_purchase_rolling_avg_3m']].copy()

purchase_lookup.rename(columns={'quantity': 'inbound_qty_lag_1m_lookup', 'unit_price': 'unit_price_lookup', 'price_vs_category_avg': 'pvc_lookup'}, inplace=True)
purchase_lookup = pd.merge(purchase_lookup, purchase_q4_stats.rename(columns={'mean': 'ira_lookup', 'max': 'irmax_lookup', 'min': 'irmin_lookup'}), on='product_name', how='left')
purchase_2024_early_lookup = monthly_purchase_hist[monthly_purchase_hist['year'] == 2024][['product_name', 'month', 'quantity']].rename(columns={'quantity': 'lag_1y_lookup'})

# ============================================================================
# 4. 2025년 1-3월 미래 데이터프레임 생성 및 값 채우기
# ============================================================================
all_products = monthly_sales_hist['product_name'].unique()
future_dates = pd.to_datetime(['2025-01-01', '2025-02-01', '2025-03-01'])

# --- 4-1. 2025년용 외부 데이터 미리 준비 ---
holidays_2025 = list(holidays.KR(years=2025).keys())
daily_2025 = pd.DataFrame(pd.date_range(start='2025-01-01', end='2025-03-31', freq='D'), columns=['date_col'])
daily_2025['is_holiday'] = ((daily_2025['date_col'].dt.dayofweek >= 5) | (daily_2025['date_col'].isin(holidays_2025))).astype(int)
holiday_count_2025 = daily_2025.resample('MS', on='date_col')['is_holiday'].sum().reset_index()
holiday_count_2025.rename(columns={'is_holiday': 'holiday_count', 'date_col':'date'}, inplace=True)
holiday_count_2025['year'] = 2025
holiday_count_2025['month'] = holiday_count_2025['date'].dt.month

external_data_2025 = holiday_count_2025.drop(columns='date')
for df in [weather, cpi_ramyeon, cpi, ccsi, trends_ramyeon, trends_cup_noodles]:
    external_data_2025 = pd.merge(external_data_2025, df[df['year'] == 2025], on=['year', 'month'], how='left')

# --- 4-2. Sales 미래 데이터 생성 ---
monthly_sales_future = pd.DataFrame(pd.MultiIndex.from_product([all_products, future_dates], names=['product_name', 'sales_date']).to_frame(index=False))
monthly_sales_future['year'] = 2025
monthly_sales_future['month'] = monthly_sales_future['sales_date'].dt.month
monthly_sales_future['quarter'] = 1

# lookup 테이블 병합 시 '용기라면' 데이터가 자동으로 채워짐
monthly_sales_future = pd.merge(monthly_sales_future, sales_lookup, on='product_name', how='left')
monthly_sales_future = pd.merge(monthly_sales_future, sales_2024_early_lookup, on=['product_name', 'month'], how='left')
monthly_sales_future.rename(columns={'sales_qty_lag_1m_lookup': 'sales_qty_lag_1m', 'lag_1y_lookup': 'sales_qty_lag_1y', 'sra_lookup': 'sales_rolling_avg_3m', 'srmax_lookup': 'sales_rolling_max_3m', 'srmin_lookup': 'sales_rolling_min_3m', 'pvc_lookup': 'price_vs_category_avg', 'unit_supply_price_lookup': 'unit_supply_price'}, inplace=True)
monthly_sales_future = pd.merge(monthly_sales_future, external_data_2025, on=['year', 'month'], how='left')
# --- [수정] 아래의 중복 병합 코드를 삭제합니다 ---
# monthly_sales_future = pd.merge(monthly_sales_future, cup_noodle_monthly_sales[cup_noodle_monthly_sales['year']==2025].drop(columns='sales_date'), on=['year', 'month'], how='left')

# --- 4-3. Purchase 미래 데이터 생성 ---
monthly_purchase_future = pd.DataFrame(pd.MultiIndex.from_product([all_products, future_dates], names=['product_name', 'date']).to_frame(index=False))
monthly_purchase_future['year'] = 2025
monthly_purchase_future['month'] = monthly_purchase_future['date'].dt.month
monthly_purchase_future['quarter'] = 1

# lookup 테이블 병합 시 '용기라면' 데이터가 자동으로 채워짐
monthly_purchase_future = pd.merge(monthly_purchase_future, purchase_lookup, on='product_name', how='left')
monthly_purchase_future = pd.merge(monthly_purchase_future, purchase_2024_early_lookup, on=['product_name', 'month'], how='left')
monthly_purchase_future.rename(columns={'inbound_qty_lag_1m_lookup': 'inbound_qty_lag_1m', 'lag_1y_lookup': 'inbound_qty_lag_1y', 'ira_lookup': 'inbound_rolling_avg_3m', 'irmax_lookup': 'inbound_rolling_max_3m', 'irmin_lookup': 'inbound_rolling_min_3m', 'pvc_lookup': 'price_vs_category_avg', 'unit_price_lookup': 'unit_price'}, inplace=True)
monthly_purchase_future = pd.merge(monthly_purchase_future, external_data_2025, on=['year', 'month'], how='left')
# --- [수정] 아래의 중복 병합 코드를 삭제합니다 ---
# monthly_purchase_future = pd.merge(monthly_purchase_future, cup_noodle_monthly_purchase[cup_noodle_monthly_purchase['year']==2025].drop(columns='date'), on=['year', 'month'], how='left')

# ============================================================================
# 5. 과거 월별 데이터와 미래 월별 데이터를 통합하여 최종 결과물 생성
# ============================================================================
sales = pd.concat([monthly_sales_hist, monthly_sales_future], ignore_index=True)
sales.sort_values(['product_name', 'sales_date'], inplace=True, ignore_index=True)
sales['sub_category'] = sales.groupby('product_name')['sub_category'].transform('ffill')
sales['price_volatility'] = sales.groupby('product_name')['unit_supply_price'].rolling(3, min_periods=1).std().reset_index(level=0, drop=True)
sales.fillna(0, inplace=True)
sales.drop(columns=['sub_category', 'supply_price', 'unit_supply_price'], inplace=True)

purchase = pd.concat([monthly_purchase_hist, monthly_purchase_future], ignore_index=True)
purchase.sort_values(['product_name', 'date'], inplace=True, ignore_index=True)
purchase['sub_category'] = purchase.groupby('product_name')['sub_category'].transform('ffill')
purchase['price_volatility'] = purchase.groupby('product_name')['unit_price'].rolling(3, min_periods=1).std().reset_index(level=0, drop=True)
purchase.fillna(0, inplace=True)
purchase.drop(columns=['sub_category', 'sales_price', 'unit_price'], inplace=True)

# ============================================================================
# 6. 최종 결과 확인
# ============================================================================
print("최종 'sales' 데이터프레임 확인 (과거 + 2025년 미래 데이터 포함):")
print(sales.tail())
print("\n최종 'purchase' 데이터프레임 확인 (과거 + 2025년 미래 데이터 포함):")
print(purchase.tail())
```

### **매출 데이터**

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder
import numpy as np
from xgboost import XGBRegressor
import lightgbm as lgb
from lightgbm import LGBMRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import pandas as pd

### ===========================================================================
### 1단계: 모델 성능 검증 (2024년 데이터 사용)
### ===========================================================================
# 목표: 2024년 데이터로 여러 모델의 성능을 검증하고 최적 모델 선택

## 1-1. 모델링을 위한 데이터 전처리
# ---------------------------------------------------------------------------
# 모델 학습을 위해 문자열(object) 데이터를 숫자형으로 변환
sales['trends_ramyeon'] = pd.to_numeric(sales['trends_ramyeon'], errors='coerce').fillna(0)
sales['trends_cup_noodles'] = pd.to_numeric(sales['trends_cup_noodles'], errors='coerce').fillna(0)

# 'product_name'을 LabelEncoder로 숫자 변환
le_sales = LabelEncoder()
sales['product_name'] = le_sales.fit_transform(sales['product_name'])

## 1-2. 검증용 학습/테스트 데이터 분리
# ---------------------------------------------------------------------------
# 2024년 기준, 학습용/검증용 데이터 분리
train_sales = sales[sales['year'] < 2024].copy()
val_sales = sales[sales['year'] == 2024].copy()

print(f"훈련 데이터 형태: {train_sales.shape}")
print(f"검증 데이터 형태: {val_sales.shape}")

## 1-3. 검증용 피처(X)와 타겟(y) 정의
# ---------------------------------------------------------------------------
X_train_sales = train_sales.drop(['quantity', 'sales_date'], axis=1)
y_train_sales = train_sales['quantity']

X_val_sales = val_sales.drop(['quantity', 'sales_date'], axis=1)
y_val_sales = val_sales['quantity']

## 1-4. 다양한 모델 학습 및 성능 비교
# ---------------------------------------------------------------------------
print("--- 2024년 실제 판매량(quantity) 통계 ---")
print(y_val_sales.describe())

# --- XGBoost ---
model_xgb_sales = XGBRegressor(
    n_estimators=1000,
    learning_rate=0.05,
    max_depth=5,
    early_stopping_rounds=10,
    random_state=42,
    n_jobs=-1
)
model_xgb_sales.fit(X_train_sales,
                    y_train_sales,
                    eval_set=[(X_val_sales, y_val_sales)],
                    verbose=0
)
y_pred_xgb_sales = model_xgb_sales.predict(X_val_sales)
mae_xgb_sales = mean_absolute_error(y_val_sales, y_pred_xgb_sales)
rmse_xgb_sales = np.sqrt(mean_squared_error(y_val_sales, y_pred_xgb_sales))
print("\n--- XGBoost 검증 성능 ---")
print(f"MAE: {mae_xgb_sales:.4f}")
print(f"RMSE: {rmse_xgb_sales:.4f}")

# --- LightGBM ---
callbacks = [lgb.early_stopping(stopping_rounds=10)]
model_lgbm_sales = LGBMRegressor(
    n_estimators=1000,
    learning_rate=0.05,
    random_state=42,
    n_jobs=-1
)
model_lgbm_sales.fit(
    X_train_sales,
    y_train_sales,
    eval_set=[(X_val_sales, y_val_sales)],
    callbacks=callbacks
)
y_pred_lgbm_sales = model_lgbm_sales.predict(X_val_sales)
mae_lgbm_sales = mean_absolute_error(y_val_sales, y_pred_lgbm_sales)
rmse_lgbm_sales = np.sqrt(mean_squared_error(y_val_sales, y_pred_lgbm_sales))
print("\n--- LightGBM 검증 성능 ---")
print(f"LightGBM MAE: {mae_lgbm_sales:.4f}")
print(f"LightGBM RMSE: {rmse_lgbm_sales:.4f}")

# --- RandomForest ---
model_rf_sales = RandomForestRegressor(
    n_estimators=100,
    max_depth=10,
    min_samples_leaf=5,
    random_state=42,
    n_jobs=-1
)
model_rf_sales.fit(X_train_sales, y_train_sales)
y_pred_rf_sales = model_rf_sales.predict(X_val_sales)
mae_rf_sales = mean_absolute_error(y_val_sales, y_pred_rf_sales)
rmse_rf_sales = np.sqrt(mean_squared_error(y_val_sales, y_pred_rf_sales))
print("\n--- RandomForest 검증 성능 ---")
print(f"RandomForest MAE: {mae_rf_sales:.4f}")
print(f"RandomForest RMSE: {rmse_rf_sales:.4f}")


### ===========================================================================
### 2단계: 최종 예측 (2025년 데이터 생성)
### ===========================================================================
# 목표: 선택된 최적 모델로 2025년 매출량 예측 결과 생성

## 2-1. 최종 학습 데이터 및 예측할 데이터 준비
# ---------------------------------------------------------------------------
# 최종 학습용 데이터 준비 (2024년 데이터 포함)
train_sales = sales[sales['year'] < 2025].copy()
# 예측 대상 데이터 준비 (2025년)
val_sales = sales[sales['year'] == 2025].copy()

print(f"\n최종 훈련 데이터 형태: {train_sales.shape}")
print(f"최종 예측 데이터 형태: {val_sales.shape}")

## 2-2. 최종 학습용 피처(X)와 타겟(y) 정의
# ---------------------------------------------------------------------------
X_train_sales = train_sales.drop(['quantity', 'sales_date'], axis=1)
y_train_sales = train_sales['quantity']

# 예측할 데이터의 피처(X) 분리 (y는 존재하지 않음)
X_val_sales = val_sales.drop(['quantity', 'sales_date'], axis=1)
y_val_sales = val_sales['quantity'] # 이 변수는 여기서는 사용되지 않음 (값이 0이므로)

## 2-3. 최종 모델 재학습 및 예측
# ---------------------------------------------------------------------------
# ==> 검증 결과, RandomForest를 최종 모델로 선택
# 성능 극대화를 위해 모든 과거 데이터로 최종 모델 재학습
print("\n--- 최종 모델(RandomForest) 재학습 시작 ---")
model_rf_sales.fit(X_train_sales, y_train_sales)
print("--- 최종 모델 재학습 완료 ---")

# 2025년 매출량 예측
y_pred_rf_sales = model_rf_sales.predict(X_val_sales)

## 2-4. 예측 결과 정리 및 출력
# ---------------------------------------------------------------------------
# 예측 결과를 보기 쉬운 형태의 테이블로 정리
sales_results_rf_2025 = val_sales[['product_name', 'year', 'month']].copy()
sales_results_rf_2025['predicted_quantity'] = y_pred_rf_sales

# product_name 원래대로 복원
sales_results_rf_2025['product_name'] = le_sales.inverse_transform(sales_results_rf_2025['product_name'])

print("\n--- 2025년 제품별 최종 매출량 예측 결과 (일부) ---")
print(sales_results_rf_2025.head())

# 월별 총 예측량 계산
monthly_sales_prediction_rf_2025 = sales_results_rf_2025.groupby(['year', 'month'])['predicted_quantity'].sum().reset_index()

print("\n--- 2025년 월별 총 매출 예측량 ---")
print(monthly_sales_prediction_rf_2025)
```
